````python
# ============================================================
# 2. FEATURE ARTIFACT AUDIT AND BASE POLICY
# ============================================================

def is_b24_b25_feature(name: str) -> bool:
    return bool(re.search(r"(?:^|_)B(?:24|25)(?:_|$)", str(name)))


def is_diagnostic_feature(name: str) -> bool:
    low = str(name).lower()
    return any(pattern in low for pattern in DIAGNOSTIC_PATTERNS)


def is_beacon_id_feature(name: str) -> bool:
    low = str(name).lower()
    return low.endswith("_id") or "_id_" in low


def is_per_beacon_rssi_std(name: str) -> bool:
    return bool(re.fullmatch(r"rssi_std_B\d{2}", str(name)))


def audit_rssi_std_artifact(data: pd.DataFrame) -> pd.DataFrame:
    std_cols = [c for c in data.columns if is_per_beacon_rssi_std(c)]
    rows = []

    if not std_cols:
        return pd.DataFrame(columns=[
            "feature", "missing_ratio", "count_zero_nonmissing",
            "count_zero_value_zero", "same_missing_mask_as_first",
        ])

    first_mask = data[std_cols[0]].isna()

    for c in std_cols:
        beacon = c.rsplit("_", 1)[-1]
        count_col = f"count_{beacon}"

        if count_col in data.columns:
            count_zero = pd.to_numeric(data[count_col], errors="coerce").fillna(0).eq(0)
            count_zero_nonmissing = int((count_zero & data[c].notna()).sum())
            count_zero_value_zero = int(
                (
                    count_zero
                    & pd.to_numeric(data[c], errors="coerce").fillna(np.nan).eq(0)
                ).sum()
            )
        else:
            count_zero_nonmissing = np.nan
            count_zero_value_zero = np.nan

        rows.append({
            "feature": c,
            "missing_ratio": float(data[c].isna().mean()),
            "count_zero_nonmissing": count_zero_nonmissing,
            "count_zero_value_zero": count_zero_value_zero,
            "same_missing_mask_as_first": bool(data[c].isna().equals(first_mask)),
        })

    return pd.DataFrame(rows)


rssi_std_artifact_audit = audit_rssi_std_artifact(df)
rssi_std_artifact_audit.to_csv(
    OUTPUT_DIR / "rssi_std_feature_artifact_audit.csv",
    index=False,
)

if len(rssi_std_artifact_audit):
    identical_mask_ratio = float(
        rssi_std_artifact_audit["same_missing_mask_as_first"].mean()
    )
    count_zero_nonmissing_total = float(
        pd.to_numeric(
            rssi_std_artifact_audit["count_zero_nonmissing"],
            errors="coerce",
        ).fillna(0).sum()
    )
    RSSI_STD_ARTIFACT_CONFIRMED = (
        identical_mask_ratio >= 0.90
        and count_zero_nonmissing_total > 0
    )
else:
    RSSI_STD_ARTIFACT_CONFIRMED = False


def build_base_feature_columns(data: pd.DataFrame) -> Tuple[List[str], pd.DataFrame]:
    selected = []
    audit = []

    for c in data.columns:
        reason = None
        low = str(c).lower()

        if c in META_COLS:
            reason = "metadata"
        elif not pd.api.types.is_numeric_dtype(data[c]):
            reason = "non_numeric"
        elif not USE_TIME_FEATURES and c in TIME_FEATURES:
            reason = "time_feature"
        elif not USE_POWER_FEATURES and (
            c in POWER_FEATURES or low.startswith("power")
        ):
            reason = "power_feature"
        elif not USE_BEACON_ID_FEATURES and is_beacon_id_feature(c):
            reason = "beacon_id"
        elif not USE_B24_B25 and is_b24_b25_feature(c):
            reason = "inactive_B24_B25"
        elif is_diagnostic_feature(c):
            reason = "diagnostic_or_label_leakage"
        elif (
            DROP_RSSI_STD_ARTIFACT
            and RSSI_STD_ARTIFACT_CONFIRMED
            and is_per_beacon_rssi_std(c)
        ):
            reason = "rssi_std_extraction_artifact"
        else:
            selected.append(c)
            reason = "candidate"

        audit.append({
            "feature": c,
            "status": reason,
            "dtype": str(data[c].dtype),
            "missing_ratio": float(data[c].isna().mean()),
            "n_unique": int(data[c].nunique(dropna=True)),
        })

    return selected, pd.DataFrame(audit)


base_feature_cols, base_feature_audit = build_base_feature_columns(df)
base_feature_audit.to_csv(
    OUTPUT_DIR / "base_feature_policy_audit.csv",
    index=False,
)

print("RSSI std artifact confirmed:", RSSI_STD_ARTIFACT_CONFIRMED)
print("Base candidate features:", len(base_feature_cols))
display(rssi_std_artifact_audit.head(30))
display(base_feature_audit["status"].value_counts().rename_axis("status").reset_index(name="count"))
````

## 3. Train-only feature engineering

### Relative RSSI

Ãƒâ€žÃ‚ÂÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœi vÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºi mÃƒÂ¡Ã‚Â»Ã¢â‚¬â€i window, feature tÃƒâ€ Ã‚Â°Ãƒâ€ Ã‚Â¡ng Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœi Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c tÃƒÆ’Ã‚Â­nh bÃƒÂ¡Ã‚ÂºÃ‚Â±ng:

```text
relative RSSI of beacon i = RSSI mean of beacon i ÃƒÂ¢Ã‹â€ Ã¢â‚¬â„¢ strongest observed RSSI mean
```

PhÃƒÆ’Ã‚Â©p biÃƒÂ¡Ã‚ÂºÃ‚Â¿n Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¢i nÃƒÆ’Ã‚Â y giÃƒÂ¡Ã‚ÂºÃ‚Â£m ÃƒÂ¡Ã‚ÂºÃ‚Â£nh hÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã…Â¸ng cÃƒÂ¡Ã‚Â»Ã‚Â§a global amplitude shift giÃƒÂ¡Ã‚Â»Ã‚Â¯a cÃƒÆ’Ã‚Â¡c ngÃƒÆ’Ã‚Â y.

### Room fingerprint

Room fingerprint chÃƒÂ¡Ã‚Â»Ã¢â‚¬Â° Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c fit tÃƒÂ¡Ã‚Â»Ã‚Â« outer-train days. Test day khÃƒÆ’Ã‚Â´ng tham gia tÃƒÂ¡Ã‚ÂºÃ‚Â¡o prototype.


````python
# ============================================================
# 3. TRAIN-ONLY FEATURE ENGINEERING
# ============================================================

def normalize_rows(values: np.ndarray, eps: float = 1e-12) -> np.ndarray:
    values = np.asarray(values, dtype=float)
    values = np.nan_to_num(values, nan=0.0, posinf=0.0, neginf=0.0)
    values = np.clip(values, 0.0, None)

    sums = values.sum(axis=1, keepdims=True)
    bad = sums[:, 0] <= eps

    if np.any(bad):
        values[bad] = 1.0 / values.shape[1]
        sums = values.sum(axis=1, keepdims=True)

    return values / np.maximum(sums, eps)


def get_freq_cols(data: pd.DataFrame) -> List[str]:
    return [
        f"freq_B{i:02d}"
        for i in range(1, 24)
        if f"freq_B{i:02d}" in data.columns
    ]


def add_relative_rssi_features(data: pd.DataFrame) -> pd.DataFrame:
    out = data.copy()

    if not USE_RELATIVE_RSSI:
        return out

    mean_cols = [
        f"rssi_mean_B{i:02d}"
        for i in range(1, 24)
        if f"rssi_mean_B{i:02d}" in out.columns
    ]

    if not mean_cols:
        return out

    matrix = out[mean_cols].apply(
        pd.to_numeric,
        errors="coerce",
    ).to_numpy(dtype=float)

    has_signal = np.isfinite(matrix).any(axis=1)
    strongest = np.full(len(out), np.nan, dtype=float)

    if has_signal.any():
        strongest[has_signal] = np.nanmax(matrix[has_signal], axis=1)

    for col_idx, source_col in enumerate(mean_cols):
        beacon = source_col.rsplit("_", 1)[-1]
        out[f"rssi_rel_{beacon}"] = matrix[:, col_idx] - strongest

    return out


def fit_room_fingerprints(
    train_data: pd.DataFrame,
    freq_cols: Sequence[str],
) -> pd.DataFrame:
    if not freq_cols:
        return pd.DataFrame()

    grouped = (
        train_data.groupby(TARGET_COL, observed=True)[list(freq_cols)]
                  .mean()
    )

    normalized = normalize_rows(grouped.to_numpy(dtype=float))

    return pd.DataFrame(
        normalized,
        index=grouped.index.astype(str),
        columns=list(freq_cols),
    )


def add_fingerprint_features(
    data: pd.DataFrame,
    fingerprints: pd.DataFrame,
    freq_cols: Sequence[str],
) -> pd.DataFrame:
    out = data.copy()

    if not USE_FINGERPRINT_FEATURES:
        return out

    if fingerprints.empty or not freq_cols:
        return out

    P = normalize_rows(
        out[list(freq_cols)].to_numpy(dtype=float)
    )
    Q = normalize_rows(
        fingerprints[list(freq_cols)].to_numpy(dtype=float)
    )

    p_norm = np.linalg.norm(P, axis=1, keepdims=True)
    q_norm = np.linalg.norm(Q, axis=1, keepdims=True).T

    cosine = (P @ Q.T) / np.maximum(
        p_norm * q_norm,
        1e-12,
    )

    js = np.empty((len(P), len(Q)), dtype=float)
    l1 = np.empty_like(js)
    log2 = np.log(2.0)

    for j in range(len(Q)):
        q = Q[j][None, :]
        midpoint = 0.5 * (P + q)

        kl_pm = np.sum(
            np.where(
                P > 0,
                P
                * (
                    np.log(np.maximum(P, 1e-12))
                    - np.log(np.maximum(midpoint, 1e-12))
                )
                / log2,
                0.0,
            ),
            axis=1,
        )

        kl_qm = np.sum(
            np.where(
                q > 0,
                q
                * (
                    np.log(np.maximum(q, 1e-12))
                    - np.log(np.maximum(midpoint, 1e-12))
                )
                / log2,
                0.0,
            ),
            axis=1,
        )

        js[:, j] = np.sqrt(
            np.maximum(0.5 * (kl_pm + kl_qm), 0.0)
        )
        l1[:, j] = np.abs(P - q).sum(axis=1)

    cosine_sorted = np.sort(cosine, axis=1)
    js_sorted = np.sort(js, axis=1)
    l1_sorted = np.sort(l1, axis=1)

    out["fp_cos_best"] = cosine_sorted[:, -1]
    out["fp_cos_second"] = (
        cosine_sorted[:, -2]
        if cosine.shape[1] > 1
        else cosine_sorted[:, -1]
    )
    out["fp_cos_margin"] = (
        out["fp_cos_best"] - out["fp_cos_second"]
    )

    out["fp_js_best"] = js_sorted[:, 0]
    out["fp_js_second"] = (
        js_sorted[:, 1]
        if js.shape[1] > 1
        else js_sorted[:, 0]
    )
    out["fp_js_margin"] = (
        out["fp_js_second"] - out["fp_js_best"]
    )

    out["fp_l1_best"] = l1_sorted[:, 0]
    out["fp_l1_second"] = (
        l1_sorted[:, 1]
        if l1.shape[1] > 1
        else l1_sorted[:, 0]
    )
    out["fp_l1_margin"] = (
        out["fp_l1_second"] - out["fp_l1_best"]
    )

    best_cos = cosine.argmax(axis=1)
    best_js = js.argmin(axis=1)

    out["fp_metric_agreement"] = (
        best_cos == best_js
    ).astype(float)

    out["fp_match_confidence"] = (
        out["fp_cos_margin"] * out["fp_js_margin"]
    )

    return out


def engineer_fold_features(
    train_data: pd.DataFrame,
    test_data: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    train_eng = add_relative_rssi_features(train_data)
    test_eng = add_relative_rssi_features(test_data)

    freq_cols = get_freq_cols(train_eng)
    fingerprints = fit_room_fingerprints(
        train_eng,
        freq_cols,
    )

    train_eng = add_fingerprint_features(
        train_eng,
        fingerprints,
        freq_cols,
    )
    test_eng = add_fingerprint_features(
        test_eng,
        fingerprints,
        freq_cols,
    )

    return train_eng, test_eng, fingerprints
````

````python
# ============================================================
# 4. DRIFT-AWARE, CORRELATION-AWARE FEATURE SELECTOR
# ============================================================

def minmax01(values: pd.Series) -> pd.Series:
    values = pd.to_numeric(values, errors="coerce").fillna(0.0)
    low = float(values.min())
    high = float(values.max())

    if high - low <= 1e-12:
        return pd.Series(
            np.zeros(len(values)),
            index=values.index,
            dtype=float,
        )

    return (values - low) / (high - low)


def is_discrete_feature(name: str) -> bool:
    low = str(name).lower()
    return (
        low.startswith("present_b")
        or low.startswith("top1_beacon")
        or low.startswith("top2_beacon")
        or low.startswith("top3_beacon")
        or is_beacon_id_feature(name)
    )


def sample_rank_rows(
    data: pd.DataFrame,
    max_rows: int,
    random_state: int,
) -> pd.DataFrame:
    if len(data) <= max_rows:
        return data.copy()

    days = sorted(data[DATE_COL].astype(str).unique())
    per_day = max(1, max_rows // max(len(days), 1))
    parts = []

    for offset, day in enumerate(days):
        part = data[data[DATE_COL].astype(str) == day]
        n_take = min(per_day, len(part))
        parts.append(
            part.sample(
                n=n_take,
                random_state=random_state + offset,
            )
        )

    sampled = pd.concat(parts, axis=0)

    if len(sampled) > max_rows:
        sampled = sampled.sample(
            n=max_rows,
            random_state=random_state,
        )

    return sampled.sort_index()


def remove_duplicate_columns(
    X: pd.DataFrame,
) -> Tuple[List[str], List[str]]:
    keep = []
    dropped = []
    seen: Dict[int, str] = {}

    for c in X.columns:
        hash_value = int(
            pd.util.hash_pandas_object(
                X[c],
                index=False,
            ).sum()
        )

        if (
            hash_value in seen
            and X[c].equals(X[seen[hash_value]])
        ):
            dropped.append(c)
        else:
            seen[hash_value] = c
            keep.append(c)

    return keep, dropped


def fit_drift_aware_selector(
    train_data: pd.DataFrame,
    candidate_cols: Sequence[str],
    top_k: int = TOP_K,
    random_state: int = RANDOM_STATE,
) -> Dict[str, object]:
    y_room = train_data[TARGET_COL].astype(str)
    y_day = train_data[DATE_COL].astype(str)

    X0 = (
        train_data.reindex(columns=list(candidate_cols))
                  .copy()
                  .replace([np.inf, -np.inf], np.nan)
    )

    audit_rows = []
    working_cols = []

    for c in X0.columns:
        missing_ratio = float(X0[c].isna().mean())
        nunique = int(X0[c].nunique(dropna=True))

        if X0[c].isna().all():
            status = "all_nan"
        elif missing_ratio > MAX_MISSING_RATIO:
            status = "excessive_missing"
        elif nunique <= 1:
            status = "constant"
        else:
            status = "working"
            working_cols.append(c)

        audit_rows.append({
            "feature": c,
            "initial_status": status,
            "missing_ratio": missing_ratio,
            "n_unique": nunique,
        })

    if not working_cols:
        raise ValueError("No usable features after missing/constant filtering.")

    medians = X0[working_cols].median(numeric_only=True)
    X_fill = (
        X0[working_cols]
        .fillna(medians)
        .fillna(0.0)
        .astype(float)
    )

    near_constant_keep = []
    near_constant_drop = []

    n_rows = max(len(X_fill), 1)

    for c in X_fill.columns:
        value_counts = X_fill[c].value_counts(
            dropna=False,
        )
        dominant_ratio = (
            float(value_counts.iloc[0] / n_rows)
            if len(value_counts)
            else 1.0
        )

        if dominant_ratio >= NEAR_CONSTANT_DOMINANCE:
            near_constant_drop.append(c)
        else:
            near_constant_keep.append(c)

    X_nc = X_fill[near_constant_keep].copy()
    duplicate_keep, duplicate_drop = remove_duplicate_columns(X_nc)
    X_clean = X_nc[duplicate_keep].copy()

    if X_clean.empty:
        raise ValueError("No features remain before ranking.")

    rank_source = train_data[
        [TARGET_COL, DATE_COL]
        + list(X_clean.columns)
    ].copy()

    rank_source = sample_rank_rows(
        rank_source,
        MAX_RANK_SAMPLES,
        random_state,
    )

    X_rank = (
        rank_source[X_clean.columns]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(medians.reindex(X_clean.columns))
        .fillna(0.0)
        .astype(float)
    )

    room_encoder = LabelEncoder()
    day_encoder = LabelEncoder()

    y_room_enc = room_encoder.fit_transform(
        rank_source[TARGET_COL].astype(str)
    )
    y_day_enc = day_encoder.fit_transform(
        rank_source[DATE_COL].astype(str)
    )

    discrete_mask = np.array(
        [is_discrete_feature(c) for c in X_clean.columns],
        dtype=bool,
    )

    mi_room = mutual_info_classif(
        X_rank,
        y_room_enc,
        discrete_features=discrete_mask,
        random_state=random_state,
    )

    if len(np.unique(y_day_enc)) > 1:
        mi_day = mutual_info_classif(
            X_rank,
            y_day_enc,
            discrete_features=discrete_mask,
            random_state=random_state,
        )
    else:
        mi_day = np.zeros(len(X_clean.columns), dtype=float)

    extra_trees = ExtraTreesClassifier(
        n_estimators=250,
        max_features="sqrt",
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=random_state,
        n_jobs=MODEL_N_JOBS,
    )
    extra_trees.fit(X_rank, y_room_enc)

    ranking = pd.DataFrame({
        "feature": X_clean.columns,
        "mi_room": mi_room,
        "mi_day": mi_day,
        "extratrees_importance": extra_trees.feature_importances_,
        "missing_ratio": [
            float(X0[c].isna().mean())
            for c in X_clean.columns
        ],
    })

    ranking["mi_room_norm"] = minmax01(ranking["mi_room"])
    ranking["mi_day_norm"] = minmax01(ranking["mi_day"])
    ranking["et_norm"] = minmax01(
        ranking["extratrees_importance"]
    )

    ranking["day_to_room_mi_ratio"] = (
        ranking["mi_day"]
        / (ranking["mi_room"] + 1e-12)
    )

    ranking["strong_day_dominant"] = (
        (ranking["mi_day"] >= DAY_MI_MIN)
        & (
            ranking["mi_day"]
            > DAY_TO_ROOM_MI_RATIO
            * ranking["mi_room"]
        )
    )

    ranking["priority_score"] = (
        0.45 * ranking["mi_room_norm"]
        + 0.45 * ranking["et_norm"]
        - DAY_MI_PENALTY * ranking["mi_day_norm"]
        - MISSING_PENALTY * ranking["missing_ratio"]
    )

    if DROP_STRONG_DAY_DOMINANT:
        eligible = ranking[
            ~ranking["strong_day_dominant"]
        ].copy()
    else:
        eligible = ranking.copy()

    if eligible.empty:
        raise ValueError(
            "All features were marked day-dominant. "
            "Reduce DAY_MI_MIN or disable hard removal."
        )

    # Spearman correlation is fit only on the outer-train pool.
    corr_source = X_clean[eligible["feature"].tolist()]

    if len(corr_source) > MAX_RANK_SAMPLES:
        corr_source = corr_source.sample(
            n=MAX_RANK_SAMPLES,
            random_state=random_state,
        )

    corr_matrix = corr_source.corr(
        method="spearman",
    ).abs()

    ordered_features = (
        eligible.sort_values(
            ["priority_score", "mi_room", "extratrees_importance"],
            ascending=[False, False, False],
        )["feature"]
        .tolist()
    )

    kept_after_corr = []
    corr_drop_rows = []

    for feature in ordered_features:
        conflict_feature = None
        conflict_value = np.nan

        for kept in kept_after_corr:
            corr_value = corr_matrix.at[feature, kept]

            if (
                np.isfinite(corr_value)
                and corr_value >= CORR_THRESHOLD
            ):
                conflict_feature = kept
                conflict_value = float(corr_value)
                break

        if conflict_feature is None:
            kept_after_corr.append(feature)
        else:
            corr_drop_rows.append({
                "dropped_feature": feature,
                "kept_feature": conflict_feature,
                "abs_spearman_corr": conflict_value,
            })

    ranking = ranking.sort_values(
        ["priority_score", "mi_room"],
        ascending=[False, False],
    ).reset_index(drop=True)

    ranking["eligible_after_day_filter"] = (
        ranking["feature"].isin(
            eligible["feature"]
        )
    )
    ranking["kept_after_corr"] = (
        ranking["feature"].isin(
            kept_after_corr
        )
    )

    selected_features = kept_after_corr[
        : min(int(top_k), len(kept_after_corr))
    ]

    ranking["selected_top_k"] = (
        ranking["feature"].isin(selected_features)
    )

    cleaning_audit = pd.DataFrame(audit_rows)
    cleaning_audit["near_constant_dropped"] = (
        cleaning_audit["feature"].isin(
            near_constant_drop
        )
    )
    cleaning_audit["duplicate_dropped"] = (
        cleaning_audit["feature"].isin(
            duplicate_drop
        )
    )

    return {
        "selected_features": selected_features,
        "ranking": ranking,
        "cleaning_audit": cleaning_audit,
        "correlation_dropped": pd.DataFrame(
            corr_drop_rows
        ),
        "medians": medians,
        "near_constant_dropped": near_constant_drop,
        "duplicate_dropped": duplicate_drop,
    }


def transform_selected(
    data: pd.DataFrame,
    selector: Dict[str, object],
) -> pd.DataFrame:
    selected = selector["selected_features"]

    return (
        data.reindex(columns=selected)
            .replace([np.inf, -np.inf], np.nan)
            .astype(float)
    )
````

````python
# ============================================================
# 5. TRAIN-ONLY AUGMENTATION
# ============================================================

def selected_rssi_jitter_columns(
    selected_features: Sequence[str],
) -> List[str]:
    columns = []

    allowed_prefixes = (
        "rssi_mean_B",
        "rssi_median_B",
        "rssi_min_B",
        "rssi_max_B",
        "rssi_rel_B",
        "rssi_mean_all",
        "rssi_median_all",
        "rssi_min_all",
        "rssi_max_all",
    )

    for c in selected_features:
        if c.startswith(allowed_prefixes):
            columns.append(c)

    return columns


def recompute_relative_rssi_inplace(
    data: pd.DataFrame,
) -> pd.DataFrame:
    if not USE_RELATIVE_RSSI:
        return data

    out = data.copy()

    mean_cols = [
        f"rssi_mean_B{i:02d}"
        for i in range(1, 24)
        if f"rssi_mean_B{i:02d}" in out.columns
    ]

    if not mean_cols:
        return out

    matrix = out[mean_cols].to_numpy(dtype=float)
    has_signal = np.isfinite(matrix).any(axis=1)
    strongest = np.full(len(out), np.nan, dtype=float)

    if has_signal.any():
        strongest[has_signal] = np.nanmax(
            matrix[has_signal],
            axis=1,
        )

    for idx, source_col in enumerate(mean_cols):
        beacon = source_col.rsplit("_", 1)[-1]
        rel_col = f"rssi_rel_{beacon}"

        if rel_col in out.columns:
            out[rel_col] = (
                matrix[:, idx] - strongest
            )

    return out


def augment_train_only(
    train_data: pd.DataFrame,
    selected_features: Sequence[str],
    random_state: int,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    if not ENABLE_AUGMENTATION:
        base = train_data.copy()
        base["_is_augmented"] = 0
        return base, pd.DataFrame()

    rng = np.random.RandomState(random_state)
    class_counts = train_data[TARGET_COL].value_counts()
    target_count = int(
        np.quantile(
            class_counts.to_numpy(),
            AUG_TARGET_QUANTILE,
        )
    )

    jitter_cols = selected_rssi_jitter_columns(
        selected_features
    )

    jitter_scale = (
        train_data[jitter_cols]
        .std(numeric_only=True)
        .replace(0, np.nan)
        .fillna(1.0)
        if jitter_cols
        else pd.Series(dtype=float)
    )

    new_parts = []
    audit_rows = []

    for room, n_original in class_counts.items():
        n_original = int(n_original)

        max_by_multiplier = max(
            int(AUG_MAX_MULTIPLIER * n_original)
            - n_original,
            0,
        )

        n_new = min(
            max(target_count - n_original, 0),
            AUG_MAX_NEW_PER_CLASS,
            max_by_multiplier,
        )

        if n_new <= 0:
            audit_rows.append({
                "room": room,
                "original_n": n_original,
                "augmented_n": 0,
                "final_n": n_original,
                "target_n": target_count,
            })
            continue

        source = train_data[
            train_data[TARGET_COL] == room
        ]

        augmented = source.sample(
            n=n_new,
            replace=True,
            random_state=random_state + len(new_parts),
        ).copy()

        if jitter_cols and RSSI_JITTER_STD_FRAC > 0:
            values = augmented[jitter_cols].to_numpy(
                dtype=float
            )

            noise_scale = (
                jitter_scale.reindex(jitter_cols)
                .to_numpy(dtype=float)
                * RSSI_JITTER_STD_FRAC
            )

            noise = rng.normal(
                loc=0.0,
                scale=noise_scale,
                size=values.shape,
            )

            valid = np.isfinite(values)
            values[valid] = values[valid] + noise[valid]
            augmented[jitter_cols] = values
            augmented = recompute_relative_rssi_inplace(
                augmented
            )

        augmented["_is_augmented"] = 1
        new_parts.append(augmented)

        audit_rows.append({
            "room": room,
            "original_n": n_original,
            "augmented_n": int(n_new),
            "final_n": int(n_original + n_new),
            "target_n": target_count,
        })

    base = train_data.copy()
    base["_is_augmented"] = 0

    if new_parts:
        augmented_train = pd.concat(
            [base] + new_parts,
            ignore_index=True,
        )
    else:
        augmented_train = base

    return augmented_train, pd.DataFrame(audit_rows)
````

````python
# ============================================================
# 6. XGBOOST PROBABILITY ENSEMBLE
# ============================================================

def normalize_probabilities(
    probabilities: np.ndarray,
    eps: float = 1e-12,
) -> np.ndarray:
    probabilities = np.asarray(
        probabilities,
        dtype=float,
    )

    probabilities = np.nan_to_num(
        probabilities,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )

    probabilities = np.clip(
        probabilities,
        0.0,
        None,
    )

    row_sums = probabilities.sum(
        axis=1,
        keepdims=True,
    )

    bad = row_sums[:, 0] <= eps

    if np.any(bad):
        probabilities[bad] = (
            1.0 / probabilities.shape[1]
        )
        row_sums = probabilities.sum(
            axis=1,
            keepdims=True,
        )

    return probabilities / np.maximum(
        row_sums,
        eps,
    )


def align_probabilities(
    probabilities: np.ndarray,
    source_classes: Sequence[str],
    target_classes: Sequence[str],
) -> np.ndarray:
    source_classes = list(map(str, source_classes))
    target_classes = list(map(str, target_classes))

    source_position = {
        c: idx for idx, c in enumerate(source_classes)
    }

    aligned = np.zeros(
        (len(probabilities), len(target_classes)),
        dtype=float,
    )

    for target_idx, class_name in enumerate(
        target_classes
    ):
        if class_name in source_position:
            aligned[:, target_idx] = probabilities[
                :,
                source_position[class_name],
            ]

    return normalize_probabilities(aligned)


def make_xgb_model(
    n_classes: int,
    seed: int,
) -> XGBClassifier:
    return XGBClassifier(
        **BEST_XGB_PARAMS,
        objective="multi:softprob",
        num_class=n_classes,
        eval_metric="mlogloss",
        tree_method="hist",
        missing=np.nan,
        random_state=seed,
        n_jobs=MODEL_N_JOBS,
    )


def fit_predict_xgb_ensemble(
    X_train: pd.DataFrame,
    y_train: Sequence[str],
    X_test: pd.DataFrame,
    target_classes: Sequence[str],
    sample_weight: Optional[np.ndarray],
    seeds: Sequence[int],
    save_dir: Optional[Path] = None,
) -> Tuple[np.ndarray, List[Dict[str, object]]]:
    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(
        pd.Series(y_train).astype(str)
    )

    source_classes = (
        label_encoder.classes_.astype(str)
    )

    probability_list = []
    model_records = []

    for seed in seeds:
        model = make_xgb_model(
            n_classes=len(source_classes),
            seed=int(seed),
        )

        fit_kwargs = {}

        if sample_weight is not None:
            fit_kwargs["sample_weight"] = sample_weight

        model.fit(
            X_train,
            y_encoded,
            **fit_kwargs,
        )

        probabilities = model.predict_proba(X_test)
        probabilities = align_probabilities(
            probabilities,
            source_classes,
            target_classes,
        )
        probability_list.append(probabilities)

        model_path = None

        if save_dir is not None:
            save_dir.mkdir(
                parents=True,
                exist_ok=True,
            )
            model_path = (
                save_dir
                / f"xgb_seed_{seed}.joblib"
            )
            joblib.dump(
                {
                    "model": model,
                    "classes": source_classes.tolist(),
                },
                model_path,
            )

        model_records.append({
            "seed": int(seed),
            "model": model,
            "source_classes": source_classes.tolist(),
            "model_path": (
                str(model_path)
                if model_path is not None
                else None
            ),
        })

    ensemble = np.mean(
        np.stack(probability_list, axis=0),
        axis=0,
    )

    return normalize_probabilities(ensemble), model_records
````

````python
# ============================================================
# 7. CAUSAL SMOOTHING, MARKOV AND VITERBI
# ============================================================

def split_contiguous_segments(
    timestamps: Sequence[pd.Timestamp],
    window_seconds: int,
) -> List[np.ndarray]:
    timestamp_series = pd.to_datetime(
        pd.Series(timestamps)
    ).reset_index(drop=True)

    if len(timestamp_series) == 0:
        return []

    max_gap = max(
        float(window_seconds) * MAX_GAP_MULTIPLIER,
        float(window_seconds) + 1.0,
    )

    gaps = (
        timestamp_series.diff()
        .dt.total_seconds()
        .fillna(0.0)
    )

    segment_id = (gaps > max_gap).cumsum()

    return [
        np.asarray(indices, dtype=int)
        for indices in segment_id.groupby(
            segment_id
        ).groups.values()
    ]


def smooth_segment_probabilities(
    probabilities: np.ndarray,
    window: int,
    mode: str,
) -> np.ndarray:
    probabilities = normalize_probabilities(
        probabilities
    )

    if window <= 1 or len(probabilities) <= 1:
        return probabilities

    if mode not in {"causal", "centered"}:
        raise ValueError(
            "SMOOTH_MODE must be causal or centered."
        )

    if window == 5:
        kernel = (
            np.array(
                [0.10, 0.15, 0.20, 0.25, 0.30],
                dtype=float,
            )
            if mode == "causal"
            else np.array(
                [0.15, 0.20, 0.30, 0.20, 0.15],
                dtype=float,
            )
        )
    else:
        kernel = np.ones(
            window,
            dtype=float,
        ) / window

    output = np.zeros_like(
        probabilities,
        dtype=float,
    )

    if mode == "causal":
        for idx in range(len(probabilities)):
            start = max(
                0,
                idx - window + 1,
            )
            segment = probabilities[start : idx + 1]
            weights = kernel[-len(segment) :]
            weights = weights / weights.sum()
            output[idx] = (
                segment * weights[:, None]
            ).sum(axis=0)

    else:
        radius = window // 2

        for idx in range(len(probabilities)):
            start = max(0, idx - radius)
            end = min(
                len(probabilities),
                idx + radius + 1,
            )

            segment = probabilities[start:end]
            kernel_start = radius - (idx - start)
            weights = kernel[
                kernel_start : kernel_start + len(segment)
            ]
            weights = weights / weights.sum()
            output[idx] = (
                segment * weights[:, None]
            ).sum(axis=0)

    return normalize_probabilities(output)


def smooth_by_segments(
    probabilities: np.ndarray,
    timestamps: Sequence[pd.Timestamp],
    window_seconds: int,
) -> np.ndarray:
    output = np.zeros_like(
        probabilities,
        dtype=float,
    )

    for indices in split_contiguous_segments(
        timestamps,
        window_seconds,
    ):
        output[indices] = smooth_segment_probabilities(
            probabilities[indices],
            window=SMOOTH_WINDOW,
            mode=SMOOTH_MODE,
        )

    return normalize_probabilities(output)


def fit_markov_transition(
    train_data: pd.DataFrame,
    classes: Sequence[str],
) -> np.ndarray:
    classes = list(map(str, classes))
    class_position = {
        c: idx for idx, c in enumerate(classes)
    }

    transition_counts = np.full(
        (len(classes), len(classes)),
        MARKOV_EPS,
        dtype=float,
    )

    max_gap = max(
        WINDOW_SECONDS * MAX_GAP_MULTIPLIER,
        WINDOW_SECONDS + 1.0,
    )

    for _, day_data in (
        train_data.sort_values(
            [DATE_COL, TIME_COL]
        ).groupby(DATE_COL)
    ):
        rooms = (
            day_data[TARGET_COL]
            .astype(str)
            .to_numpy()
        )
        timestamps = pd.to_datetime(
            day_data[TIME_COL]
        ).to_numpy()

        for idx in range(1, len(day_data)):
            gap = (
                pd.Timestamp(timestamps[idx])
                - pd.Timestamp(timestamps[idx - 1])
            ).total_seconds()

            if gap > max_gap:
                continue

            previous_room = rooms[idx - 1]
            current_room = rooms[idx]

            if (
                previous_room in class_position
                and current_room in class_position
            ):
                transition_counts[
                    class_position[previous_room],
                    class_position[current_room],
                ] += 1.0

    return transition_counts / transition_counts.sum(
        axis=1,
        keepdims=True,
    )


def viterbi_decode(
    emissions: np.ndarray,
    transition: np.ndarray,
    markov_lambda: float = MARKOV_LAMBDA,
) -> np.ndarray:
    emissions = normalize_probabilities(emissions)
    n_steps, n_classes = emissions.shape

    if n_steps == 0:
        return np.array([], dtype=int)

    log_emissions = np.log(
        np.clip(emissions, 1e-12, 1.0)
    )
    log_transition = (
        markov_lambda
        * np.log(
            np.clip(transition, 1e-12, 1.0)
        )
    )

    dynamic = np.full(
        (n_steps, n_classes),
        -np.inf,
        dtype=float,
    )
    backpointer = np.zeros(
        (n_steps, n_classes),
        dtype=int,
    )

    dynamic[0] = (
        log_emissions[0]
        - math.log(n_classes)
    )

    for step in range(1, n_steps):
        scores = (
            dynamic[step - 1][:, None]
            + log_transition
        )

        backpointer[step] = scores.argmax(axis=0)
        dynamic[step] = (
            scores.max(axis=0)
            + log_emissions[step]
        )

    path = np.zeros(n_steps, dtype=int)
    path[-1] = int(dynamic[-1].argmax())

    for step in range(n_steps - 2, -1, -1):
        path[step] = backpointer[
            step + 1,
            path[step + 1],
        ]

    return path


def viterbi_by_segments(
    probabilities: np.ndarray,
    timestamps: Sequence[pd.Timestamp],
    transition: np.ndarray,
) -> np.ndarray:
    path = np.zeros(
        len(probabilities),
        dtype=int,
    )

    for indices in split_contiguous_segments(
        timestamps,
        WINDOW_SECONDS,
    ):
        path[indices] = viterbi_decode(
            probabilities[indices],
            transition,
            markov_lambda=MARKOV_LAMBDA,
        )

    return path
````

````python
# ============================================================
# 8. METRICS, CONFUSION AND DRIFT DIAGNOSTICS
# ============================================================

def boundary_mask(
    y_true: Sequence[str],
    radius: int = 2,
) -> np.ndarray:
    labels = np.asarray(
        pd.Series(y_true).astype(str)
    )

    changes = np.zeros(
        len(labels),
        dtype=bool,
    )

    if len(labels) > 1:
        changes[1:] |= labels[1:] != labels[:-1]
        changes[:-1] |= labels[:-1] != labels[1:]

    mask = changes.copy()

    for offset in range(1, radius + 1):
        mask[offset:] |= changes[:-offset]
        mask[:-offset] |= changes[offset:]

    return mask


def evaluate_prediction(
    y_true: Sequence[str],
    y_pred: Sequence[str],
    train_classes: Sequence[str],
    stage: str,
) -> pd.DataFrame:
    y_true = pd.Series(y_true).astype(str).to_numpy()
    y_pred = pd.Series(y_pred).astype(str).to_numpy()
    train_class_set = set(map(str, train_classes))

    rows = []

    protocol_masks = {
        "open_set": np.ones(
            len(y_true),
            dtype=bool,
        ),
        "closed_set": np.array(
            [label in train_class_set for label in y_true],
            dtype=bool,
        ),
    }

    for protocol, mask in protocol_masks.items():
        if not mask.any():
            continue

        true_subset = y_true[mask]
        pred_subset = y_pred[mask]
        labels = sorted(pd.unique(true_subset))

        rows.append({
            "stage": stage,
            "protocol": protocol,
            "n_samples": int(mask.sum()),
            "n_labels": int(len(labels)),
            "accuracy": float(
                accuracy_score(
                    true_subset,
                    pred_subset,
                )
            ),
            "balanced_accuracy": float(
                balanced_accuracy_score(
                    true_subset,
                    pred_subset,
                )
            ),
            "macro_f1": float(
                f1_score(
                    true_subset,
                    pred_subset,
                    labels=labels,
                    average="macro",
                    zero_division=0,
                )
            ),
            "weighted_f1": float(
                f1_score(
                    true_subset,
                    pred_subset,
                    labels=labels,
                    average="weighted",
                    zero_division=0,
                )
            ),
            "unseen_support": int(
                sum(
                    label not in train_class_set
                    for label in true_subset
                )
            ),
        })

    return pd.DataFrame(rows)


def save_confusion_outputs(
    y_true: Sequence[str],
    y_pred: Sequence[str],
    output_stem: Path,
) -> None:
    y_true = pd.Series(y_true).astype(str)
    y_pred = pd.Series(y_pred).astype(str)

    labels = sorted(
        set(y_true.unique())
        | set(y_pred.unique())
    )

    matrix = confusion_matrix(
        y_true,
        y_pred,
        labels=labels,
    )

    matrix_df = pd.DataFrame(
        matrix,
        index=labels,
        columns=labels,
    )
    matrix_df.to_csv(
        output_stem.with_suffix(".csv")
    )

    row_sums = matrix.sum(
        axis=1,
        keepdims=True,
    )
    normalized = np.divide(
        matrix,
        row_sums,
        out=np.zeros_like(matrix, dtype=float),
        where=row_sums > 0,
    )

    figure_size = max(
        9,
        min(18, 0.65 * len(labels)),
    )

    plt.figure(
        figsize=(figure_size, figure_size)
    )
    image = plt.imshow(
        normalized,
        aspect="auto",
        vmin=0,
        vmax=1,
    )
    plt.colorbar(
        image,
        fraction=0.046,
        pad=0.04,
    )
    plt.xticks(
        np.arange(len(labels)),
        labels,
        rotation=90,
        fontsize=8,
    )
    plt.yticks(
        np.arange(len(labels)),
        labels,
        fontsize=8,
    )
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(
        f"Row-normalized confusion matrix ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â {output_stem.name}"
    )
    plt.tight_layout()
    plt.savefig(
        output_stem.with_suffix(".png"),
        dpi=180,
        bbox_inches="tight",
    )
    plt.close()


def top_confusions(
    y_true: Sequence[str],
    y_pred: Sequence[str],
    top_n: int = 40,
) -> pd.DataFrame:
    errors = pd.DataFrame({
        "true": pd.Series(y_true).astype(str),
        "pred": pd.Series(y_pred).astype(str),
    })

    errors = errors[
        errors["true"] != errors["pred"]
    ]

    if errors.empty:
        return pd.DataFrame(
            columns=["true", "pred", "count"]
        )

    return (
        errors.value_counts(
            ["true", "pred"]
        )
        .reset_index(name="count")
        .head(top_n)
    )


def feature_drift_table(
    train_data: pd.DataFrame,
    test_data: pd.DataFrame,
    selected_features: Sequence[str],
) -> pd.DataFrame:
    rows = []

    for c in selected_features:
        train_values = pd.to_numeric(
            train_data[c],
            errors="coerce",
        )
        test_values = pd.to_numeric(
            test_data[c],
            errors="coerce",
        )

        train_mean = float(train_values.mean())
        test_mean = float(test_values.mean())
        train_std = float(train_values.std())

        rows.append({
            "feature": c,
            "train_mean": train_mean,
            "test_mean": test_mean,
            "normalized_mean_shift": (
                abs(test_mean - train_mean)
                / max(
                    train_std
                    if np.isfinite(train_std)
                    else 0.0,
                    1e-6,
                )
            ),
            "train_missing_rate": float(
                train_values.isna().mean()
            ),
            "test_missing_rate": float(
                test_values.isna().mean()
            ),
            "missing_rate_delta": float(
                test_values.isna().mean()
                - train_values.isna().mean()
            ),
        })

    return pd.DataFrame(rows).sort_values(
        "normalized_mean_shift",
        ascending=False,
    )


def fingerprint_drift_table(
    train_data: pd.DataFrame,
    test_data: pd.DataFrame,
) -> pd.DataFrame:
    freq_cols = get_freq_cols(train_data)
    common_rooms = sorted(
        set(train_data[TARGET_COL].astype(str))
        & set(test_data[TARGET_COL].astype(str))
    )

    rows = []

    for room in common_rooms:
        train_fp = (
            train_data.loc[
                train_data[TARGET_COL].astype(str) == room,
                freq_cols,
            ]
            .mean()
            .to_numpy(dtype=float)
        )
        test_fp = (
            test_data.loc[
                test_data[TARGET_COL].astype(str) == room,
                freq_cols,
            ]
            .mean()
            .to_numpy(dtype=float)
        )

        train_fp = normalize_rows(
            train_fp[None, :]
        )[0]
        test_fp = normalize_rows(
            test_fp[None, :]
        )[0]

        rows.append({
            "room": room,
            "train_support": int(
                (
                    train_data[TARGET_COL].astype(str)
                    == room
                ).sum()
            ),
            "test_support": int(
                (
                    test_data[TARGET_COL].astype(str)
                    == room
                ).sum()
            ),
            "js_distance": float(
                jensenshannon(
                    train_fp,
                    test_fp,
                    base=2.0,
                )
            ),
            "l1_distance": float(
                np.abs(train_fp - test_fp).sum()
            ),
            "cosine_similarity": float(
                np.dot(train_fp, test_fp)
                / max(
                    np.linalg.norm(train_fp)
                    * np.linalg.norm(test_fp),
                    1e-12,
                )
            ),
        })

    if not rows:
        return pd.DataFrame()

    return pd.DataFrame(rows).sort_values(
        "js_distance",
        ascending=False,
    )


def subset_diagnostics(
    y_true: np.ndarray,
    y_pred: np.ndarray,
    boundary: np.ndarray,
    stage: str,
) -> pd.DataFrame:
    rows = []

    subset_masks = {
        "boundary": boundary,
        "stable": ~boundary,
        "hallway_true": y_true == "hallway",
        "non_hallway_true": y_true != "hallway",
    }

    for subset_name, mask in subset_masks.items():
        if not mask.any():
            continue

        labels = sorted(pd.unique(y_true[mask]))

        rows.append({
            "stage": stage,
            "subset": subset_name,
            "n_samples": int(mask.sum()),
            "accuracy": float(
                accuracy_score(
                    y_true[mask],
                    y_pred[mask],
                )
            ),
            "macro_f1": float(
                f1_score(
                    y_true[mask],
                    y_pred[mask],
                    labels=labels,
                    average="macro",
                    zero_division=0,
                )
            ),
        })

    return pd.DataFrame(rows)
````

````python
# ============================================================
# 9. ONE OUTER LODO FOLD
# ============================================================

@dataclass
class FoldArtifact:
    test_day: str
    fold_dir: str
    selected_features: List[str]
    classes: List[str]
    first_model_path: Optional[str]


def run_outer_fold(
    full_data: pd.DataFrame,
    test_day: str,
) -> Tuple[pd.DataFrame, FoldArtifact]:
    fold_dir = OUTPUT_DIR / f"fold_{test_day}"
    fold_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    train_raw = (
        full_data[
            full_data[DATE_COL].astype(str)
            != str(test_day)
        ]
        .copy()
        .sort_values([DATE_COL, TIME_COL])
        .reset_index(drop=True)
    )

    test_raw = (
        full_data[
            full_data[DATE_COL].astype(str)
            == str(test_day)
        ]
        .copy()
        .sort_values([TIME_COL])
        .reset_index(drop=True)
    )

    train_classes = sorted(
        train_raw[TARGET_COL].astype(str).unique()
    )
    test_classes = sorted(
        test_raw[TARGET_COL].astype(str).unique()
    )
    unseen_classes = sorted(
        set(test_classes) - set(train_classes)
    )

    train_engineered, test_engineered, fingerprints = (
        engineer_fold_features(
            train_raw,
            test_raw,
        )
    )

    fingerprints.to_csv(
        fold_dir / "room_fingerprints_train_only.csv"
    )

    candidate_cols, fold_policy_audit = (
        build_base_feature_columns(
            train_engineered
        )
    )

    selector = fit_drift_aware_selector(
        train_engineered,
        candidate_cols,
        top_k=TOP_K,
        random_state=RANDOM_STATE,
    )

    selector["ranking"].to_csv(
        fold_dir / "feature_ranking_drift_aware.csv",
        index=False,
    )
    selector["cleaning_audit"].to_csv(
        fold_dir / "feature_cleaning_audit.csv",
        index=False,
    )
    selector["correlation_dropped"].to_csv(
        fold_dir / "high_correlation_dropped.csv",
        index=False,
    )
    fold_policy_audit.to_csv(
        fold_dir / "feature_policy_audit.csv",
        index=False,
    )

    pd.DataFrame({
        "feature": selector["selected_features"],
    }).to_csv(
        fold_dir / "selected_features.csv",
        index=False,
    )

    train_augmented, augmentation_audit = (
        augment_train_only(
            train_engineered,
            selector["selected_features"],
            random_state=RANDOM_STATE,
        )
    )
    augmentation_audit.to_csv(
        fold_dir / "augmentation_audit.csv",
        index=False,
    )

    X_train = transform_selected(
        train_augmented,
        selector,
    )
    X_test = transform_selected(
        test_engineered,
        selector,
    )

    y_train = (
        train_augmented[TARGET_COL]
        .astype(str)
        .to_numpy()
    )
    y_test = (
        test_engineered[TARGET_COL]
        .astype(str)
        .to_numpy()
    )

    model_classes = sorted(pd.unique(y_train))

    sample_weight = compute_sample_weight(
        class_weight="balanced",
        y=y_train,
    )

    probabilities_raw, model_records = (
        fit_predict_xgb_ensemble(
            X_train,
            y_train,
            X_test,
            target_classes=model_classes,
            sample_weight=sample_weight,
            seeds=SEEDS,
            save_dir=fold_dir / "models",
        )
    )

    probabilities_smooth = smooth_by_segments(
        probabilities_raw,
        test_engineered[TIME_COL],
        WINDOW_SECONDS,
    )

    transition = fit_markov_transition(
        train_raw,
        model_classes,
    )
    pd.DataFrame(
        transition,
        index=model_classes,
        columns=model_classes,
    ).to_csv(
        fold_dir / "markov_transition_train_only.csv"
    )

    prediction_indices = {
        "argmax_raw": probabilities_raw.argmax(axis=1),
        "prob_smooth": probabilities_smooth.argmax(axis=1),
        "viterbi_markov": viterbi_by_segments(
            probabilities_smooth,
            test_engineered[TIME_COL],
            transition,
        ),
    }

    class_array = np.asarray(
        model_classes,
        dtype=object,
    )

    metrics_parts = []
    subset_parts = []

    predictions = test_engineered[
        [TIME_COL, DATE_COL, TARGET_COL]
    ].copy()

    predictions = predictions.rename(
        columns={TARGET_COL: "y_true"}
    )
    predictions["is_unseen_test_class"] = (
        ~predictions["y_true"].isin(
            train_classes
        )
    )
    predictions["confidence_raw"] = (
        probabilities_raw.max(axis=1)
    )
    predictions["confidence_smooth"] = (
        probabilities_smooth.max(axis=1)
    )

    boundary = boundary_mask(
        predictions["y_true"],
        radius=2,
    )
    predictions["is_boundary_gt"] = boundary

    for stage, indices in prediction_indices.items():
        y_pred = class_array[indices]
        predictions[f"y_pred_{stage}"] = y_pred

        stage_metrics = evaluate_prediction(
            y_test,
            y_pred,
            train_classes=train_classes,
            stage=stage,
        )

        stage_metrics["test_day"] = str(test_day)
        stage_metrics["n_train"] = len(train_raw)
        stage_metrics["n_train_augmented"] = len(
            train_augmented
        )
        stage_metrics["n_test"] = len(test_raw)
        stage_metrics["n_train_classes"] = len(
            train_classes
        )
        stage_metrics["n_test_classes"] = len(
            test_classes
        )
        stage_metrics["unseen_classes"] = "|".join(
            unseen_classes
        )
        stage_metrics["n_candidates"] = len(
            candidate_cols
        )
        stage_metrics["n_selected_features"] = len(
            selector["selected_features"]
        )
        stage_metrics["n_corr_dropped"] = len(
            selector["correlation_dropped"]
        )
        stage_metrics["n_day_dominant_dropped"] = int(
            selector["ranking"][
                "strong_day_dominant"
            ].sum()
        )
        metrics_parts.append(stage_metrics)

        classification_report_df = pd.DataFrame(
            classification_report(
                y_test,
                y_pred,
                labels=sorted(pd.unique(y_test)),
                output_dict=True,
                zero_division=0,
            )
        ).T.reset_index().rename(
            columns={"index": "class"}
        )

        classification_report_df.to_csv(
            fold_dir
            / f"classification_report_{stage}.csv",
            index=False,
        )

        save_confusion_outputs(
            y_test,
            y_pred,
            fold_dir / f"confusion_{stage}",
        )

        top_confusions(
            y_test,
            y_pred,
        ).to_csv(
            fold_dir
            / f"top_confusions_{stage}.csv",
            index=False,
        )

        stage_subset = subset_diagnostics(
            y_test,
            y_pred,
            boundary,
            stage,
        )
        stage_subset["test_day"] = str(test_day)
        subset_parts.append(stage_subset)

    predictions.to_csv(
        fold_dir / "predictions.csv",
        index=False,
    )

    metrics_df = pd.concat(
        metrics_parts,
        ignore_index=True,
    )
    metrics_df.to_csv(
        fold_dir / "metrics.csv",
        index=False,
    )

    subset_df = pd.concat(
        subset_parts,
        ignore_index=True,
    )
    subset_df.to_csv(
        fold_dir / "subset_diagnostics.csv",
        index=False,
    )

    feature_drift_table(
        train_engineered,
        test_engineered,
        selector["selected_features"],
    ).to_csv(
        fold_dir / "selected_feature_drift.csv",
        index=False,
    )

    fingerprint_drift_table(
        train_raw,
        test_raw,
    ).to_csv(
        fold_dir / "room_fingerprint_drift.csv",
        index=False,
    )

    metadata = {
        "test_day": str(test_day),
        "window_seconds": WINDOW_SECONDS,
        "train_classes": train_classes,
        "test_classes": test_classes,
        "unseen_classes": unseen_classes,
        "selected_features": selector["selected_features"],
        "selected_feature_count": len(
            selector["selected_features"]
        ),
        "correlation_threshold": CORR_THRESHOLD,
        "top_k": TOP_K,
        "seeds": SEEDS,
        "xgb_params": BEST_XGB_PARAMS,
        "rssi_std_artifact_confirmed": (
            RSSI_STD_ARTIFACT_CONFIRMED
        ),
        "use_relative_rssi": USE_RELATIVE_RSSI,
        "use_fingerprint_features": (
            USE_FINGERPRINT_FEATURES
        ),
        "augmentation_enabled": (
            ENABLE_AUGMENTATION
        ),
        "smooth_window": SMOOTH_WINDOW,
        "smooth_mode": SMOOTH_MODE,
        "markov_lambda": MARKOV_LAMBDA,
    }

    (
        fold_dir / "run_metadata.json"
    ).write_text(
        json.dumps(
            metadata,
            indent=2,
            ensure_ascii=False,
        ),
        encoding="utf-8",
    )

    first_model_path = (
        model_records[0]["model_path"]
        if model_records
        else None
    )

    artifact = FoldArtifact(
        test_day=str(test_day),
        fold_dir=str(fold_dir),
        selected_features=list(
            selector["selected_features"]
        ),
        classes=model_classes,
        first_model_path=first_model_path,
    )

    return metrics_df, artifact
````

````python
# ============================================================
# 12. SHAP ACROSS FOLDS
# ============================================================

def compute_shap_for_fold(
    artifact: FoldArtifact,
) -> Optional[pd.DataFrame]:
    if (
        not RUN_SHAP
        or not HAS_SHAP
        or not artifact.first_model_path
    ):
        return None

    fold_dir = Path(artifact.fold_dir)
    payload = joblib.load(
        artifact.first_model_path
    )
    model = payload["model"]

    train_raw = df[
        df[DATE_COL].astype(str)
        != artifact.test_day
    ].copy()

    test_raw = df[
        df[DATE_COL].astype(str)
        == artifact.test_day
    ].copy()

    train_engineered, test_engineered, _ = (
        engineer_fold_features(
            train_raw,
            test_raw,
        )
    )

    X = (
        test_engineered.reindex(
            columns=artifact.selected_features
        )
        .replace([np.inf, -np.inf], np.nan)
        .astype(float)
    )

    if len(X) > SHAP_MAX_SAMPLES:
        X = X.sample(
            n=SHAP_MAX_SAMPLES,
            random_state=RANDOM_STATE,
        )

    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X)

    if isinstance(shap_values, list):
        array = np.stack(
            [np.asarray(v) for v in shap_values],
            axis=0,
        )
        mean_abs = np.abs(array).mean(
            axis=(0, 1),
        )
    else:
        array = np.asarray(shap_values)

        if array.ndim == 3:
            if array.shape[1] == X.shape[1]:
                mean_abs = np.abs(array).mean(
                    axis=(0, 2),
                )
            elif array.shape[2] == X.shape[1]:
                mean_abs = np.abs(array).mean(
                    axis=(0, 1),
                )
            else:
                raise ValueError(
                    f"Unexpected SHAP shape: {array.shape}"
                )
        elif array.ndim == 2:
            mean_abs = np.abs(array).mean(axis=0)
        else:
            raise ValueError(
                f"Unexpected SHAP ndim: {array.ndim}"
            )

    importance = pd.DataFrame({
        "feature": X.columns,
        "mean_abs_shap": mean_abs,
        "test_day": artifact.test_day,
    }).sort_values(
        "mean_abs_shap",
        ascending=False,
    )

    importance.to_csv(
        fold_dir / "shap_importance.csv",
        index=False,
    )

    top = importance.head(
        SHAP_TOP_N
    ).sort_values("mean_abs_shap")

    plt.figure(
        figsize=(10, max(6, 0.3 * len(top)))
    )
    plt.barh(
        top["feature"],
        top["mean_abs_shap"],
    )
    plt.xlabel("Mean |SHAP|")
    plt.title(
        f"SHAP feature importance ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â test {artifact.test_day}"
    )
    plt.tight_layout()
    plt.savefig(
        fold_dir / "shap_summary_bar.png",
        dpi=180,
        bbox_inches="tight",
    )
    plt.close()

    return importance


shap_tables = []

if RUN_SHAP and HAS_SHAP:
    for artifact in fold_artifacts:
        print("SHAP fold:", artifact.test_day)
        result = compute_shap_for_fold(artifact)

        if result is not None:
            shap_tables.append(result)

    if shap_tables:
        shap_all = pd.concat(
            shap_tables,
            ignore_index=True,
        )
        shap_all.to_csv(
            OUTPUT_DIR / "shap_all_folds.csv",
            index=False,
        )

        shap_stability = (
            shap_all.groupby("feature")[
                "mean_abs_shap"
            ]
            .agg(["mean", "std", "count"])
            .sort_values(
                "mean",
                ascending=False,
            )
            .reset_index()
        )
        shap_stability.to_csv(
            OUTPUT_DIR / "shap_stability.csv",
            index=False,
        )
        display(shap_stability.head(40))
else:
    print(
        "SHAP skipped. RUN_SHAP=",
        RUN_SHAP,
        "| HAS_SHAP=",
        HAS_SHAP,
    )
````

````python
# ============================================================
# 13. TRAIN FINAL MODEL ON ALL FOUR DAYS
# ============================================================

def train_final_model(
    full_data: pd.DataFrame,
) -> Dict[str, object]:
    full_engineered = add_relative_rssi_features(
        full_data
    )

    freq_cols = get_freq_cols(full_engineered)
    fingerprints = fit_room_fingerprints(
        full_engineered,
        freq_cols,
    )
    full_engineered = add_fingerprint_features(
        full_engineered,
        fingerprints,
        freq_cols,
    )

    candidate_cols, final_policy_audit = (
        build_base_feature_columns(
            full_engineered
        )
    )

    selector = fit_drift_aware_selector(
        full_engineered,
        candidate_cols,
        top_k=TOP_K,
        random_state=RANDOM_STATE,
    )

    augmented, augmentation_audit = (
        augment_train_only(
            full_engineered,
            selector["selected_features"],
            random_state=RANDOM_STATE,
        )
    )

    X_train = transform_selected(
        augmented,
        selector,
    )
    y_train = (
        augmented[TARGET_COL]
        .astype(str)
        .to_numpy()
    )
    classes = sorted(pd.unique(y_train))

    sample_weight = compute_sample_weight(
        class_weight="balanced",
        y=y_train,
    )

    # Fit final models and obtain probabilities on a one-row placeholder
    # only because the shared function also performs predict_proba.
    X_placeholder = X_train.iloc[:1].copy()

    _, model_records = fit_predict_xgb_ensemble(
        X_train,
        y_train,
        X_placeholder,
        target_classes=classes,
        sample_weight=sample_weight,
        seeds=SEEDS,
        save_dir=FINAL_MODEL_DIR / "models",
    )

    fingerprints.to_csv(
        FINAL_MODEL_DIR
        / "room_fingerprints_full_train.csv"
    )
    final_policy_audit.to_csv(
        FINAL_MODEL_DIR
        / "feature_policy_audit.csv",
        index=False,
    )
    selector["ranking"].to_csv(
        FINAL_MODEL_DIR
        / "feature_ranking_drift_aware.csv",
        index=False,
    )
    selector["cleaning_audit"].to_csv(
        FINAL_MODEL_DIR
        / "feature_cleaning_audit.csv",
        index=False,
    )
    selector["correlation_dropped"].to_csv(
        FINAL_MODEL_DIR
        / "high_correlation_dropped.csv",
        index=False,
    )
    augmentation_audit.to_csv(
        FINAL_MODEL_DIR
        / "augmentation_audit.csv",
        index=False,
    )

    pd.DataFrame({
        "feature": selector["selected_features"],
    }).to_csv(
        FINAL_MODEL_DIR
        / "selected_features.csv",
        index=False,
    )

    final_metadata = {
        "window_seconds": WINDOW_SECONDS,
        "n_original_rows": int(len(full_data)),
        "n_augmented_rows": int(len(augmented)),
        "classes": classes,
        "selected_features": selector[
            "selected_features"
        ],
        "selected_feature_count": len(
            selector["selected_features"]
        ),
        "seeds": SEEDS,
        "xgb_params": BEST_XGB_PARAMS,
        "correlation_threshold": CORR_THRESHOLD,
        "drop_rssi_std_artifact": (
            DROP_RSSI_STD_ARTIFACT
        ),
        "rssi_std_artifact_confirmed": (
            RSSI_STD_ARTIFACT_CONFIRMED
        ),
        "drop_strong_day_dominant": (
            DROP_STRONG_DAY_DOMINANT
        ),
        "use_relative_rssi": USE_RELATIVE_RSSI,
        "use_fingerprint_features": (
            USE_FINGERPRINT_FEATURES
        ),
        "smooth_window": SMOOTH_WINDOW,
        "smooth_mode": SMOOTH_MODE,
        "markov_lambda": MARKOV_LAMBDA,
        "model_paths": [
            record["model_path"]
            for record in model_records
        ],
    }

    (
        FINAL_MODEL_DIR / "final_model_metadata.json"
    ).write_text(
        json.dumps(
            final_metadata,
            indent=2,
            ensure_ascii=False,
        ),
        encoding="utf-8",
    )

    transition = fit_markov_transition(
        full_data,
        classes,
    )
    pd.DataFrame(
        transition,
        index=classes,
        columns=classes,
    ).to_csv(
        FINAL_MODEL_DIR
        / "markov_transition_full_train.csv"
    )

    return final_metadata


if RUN_FINAL_TRAIN:
    final_model_metadata = train_final_model(df)
    print(
        json.dumps(
            final_model_metadata,
            indent=2,
            ensure_ascii=False,
        )
    )
else:
    print("Final training disabled.")
````

````python
# ============================================================
# BLOCK ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â CONFUSION MATRIX THEO TÃƒÂ¡Ã‚Â»Ã‚ÂªNG NGÃƒÆ’Ã¢â€šÂ¬Y / STAGE / PROTOCOL
# DÃƒÆ’Ã‚Â¹ng sau khi Ãƒâ€žÃ¢â‚¬ËœÃƒÆ’Ã‚Â£ chÃƒÂ¡Ã‚ÂºÃ‚Â¡y xong 4-fold LODO.
#
# Ãƒâ€žÃ‚ÂÃƒÂ¡Ã‚Â»Ã‚Âc:
#   OUTPUT_DIR/fold_<date>/predictions.csv
#
# VÃƒÂ¡Ã‚ÂºÃ‚Â½ riÃƒÆ’Ã‚Âªng:
#   - argmax_raw
#   - prob_smooth
#   - viterbi_markov
#   - open_set
#   - closed_set
#   - raw count
#   - normalized theo true class
# ============================================================

from pathlib import Path
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix


# ------------------------------------------------------------
# Config
# ------------------------------------------------------------
CF_PROTOCOLS = ["open_set", "closed_set"]

# None: tÃƒÂ¡Ã‚Â»Ã‚Â± Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢ng lÃƒÂ¡Ã‚ÂºÃ‚Â¥y tÃƒÂ¡Ã‚ÂºÃ‚Â¥t cÃƒÂ¡Ã‚ÂºÃ‚Â£ cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t y_pred_* trong predictions.csv
CF_STAGES = None

# False: count matrix
# True: normalize theo tÃƒÂ¡Ã‚Â»Ã‚Â«ng true class
CF_NORMALIZATION_MODES = [False, True]

CF_FIG_DPI = 220
CF_ANNOTATE = True
CF_ANNOTATE_MAX_CLASSES = 25

CF_OUTPUT_DIR = OUTPUT_DIR / "confusion_matrices"
CF_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def cf_sort_label_key(label):
    """
    SÃƒÂ¡Ã‚ÂºÃ‚Â¯p xÃƒÂ¡Ã‚ÂºÃ‚Â¿p:
    501, 502, ... trÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºc;
    cafeteria, hallway, kitchen, nurse station... sau.
    """
    value = str(label).strip()

    try:
        return (0, int(value))
    except ValueError:
        return (1, value.lower())


def cf_normalize_rows(matrix):
    """
    ChuÃƒÂ¡Ã‚ÂºÃ‚Â©n hÃƒÆ’Ã‚Â³a mÃƒÂ¡Ã‚Â»Ã¢â‚¬â€i hÃƒÆ’Ã‚Â ng theo support cÃƒÂ¡Ã‚Â»Ã‚Â§a true class.
    """
    matrix = np.asarray(matrix, dtype=np.float64)

    row_sum = matrix.sum(axis=1, keepdims=True)

    return np.divide(
        matrix,
        row_sum,
        out=np.zeros_like(matrix, dtype=np.float64),
        where=row_sum > 0,
    )


def cf_get_protocol_mask(predictions, protocol):
    """
    open_set:
        dÃƒÆ’Ã‚Â¹ng toÃƒÆ’Ã‚Â n bÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢ test samples.

    closed_set:
        loÃƒÂ¡Ã‚ÂºÃ‚Â¡i cÃƒÆ’Ã‚Â¡c sample cÃƒÆ’Ã‚Â³ true class khÃƒÆ’Ã‚Â´ng xuÃƒÂ¡Ã‚ÂºÃ‚Â¥t hiÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡n trong train fold.
    """
    if protocol == "open_set":
        return np.ones(len(predictions), dtype=bool)

    if protocol == "closed_set":
        if "is_unseen_test_class" not in predictions.columns:
            raise ValueError(
                "ThiÃƒÂ¡Ã‚ÂºÃ‚Â¿u cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t 'is_unseen_test_class' trong predictions.csv. "
                "KhÃƒÆ’Ã‚Â´ng thÃƒÂ¡Ã‚Â»Ã†â€™ xÃƒÆ’Ã‚Â¡c Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¹nh closed-set."
            )

        unseen = predictions["is_unseen_test_class"]

        if unseen.dtype == bool:
            unseen_mask = unseen.to_numpy()
        else:
            unseen_mask = (
                unseen.astype(str)
                .str.lower()
                .isin(["true", "1", "yes"])
                .to_numpy()
            )

        return ~unseen_mask

    raise ValueError(f"Unknown protocol: {protocol}")


def cf_plot_one_matrix(
    y_true,
    y_pred,
    labels,
    test_day,
    stage,
    protocol,
    normalize,
    output_dir,
):
    """
    TÃƒÂ¡Ã‚ÂºÃ‚Â¡o mÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t figure Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢c lÃƒÂ¡Ã‚ÂºÃ‚Â­p vÃƒÆ’Ã‚Â  lÃƒâ€ Ã‚Â°u matrix CSV.
    """
    matrix_count = confusion_matrix(
        y_true,
        y_pred,
        labels=labels,
    )

    if normalize:
        matrix_plot = cf_normalize_rows(matrix_count)
        value_format = ".2f"
        matrix_type = "normalized_true"
    else:
        matrix_plot = matrix_count
        value_format = "d"
        matrix_type = "count"

    matrix_df = pd.DataFrame(
        matrix_plot,
        index=labels,
        columns=labels,
    )
    matrix_df.index.name = "true"
    matrix_df.columns.name = "predicted"

    csv_path = (
        output_dir
        / f"cf_{test_day}_{protocol}_{stage}_{matrix_type}.csv"
    )
    matrix_df.to_csv(csv_path)

    n_classes = len(labels)
    figure_size = max(
        8,
        min(20, 0.72 * n_classes),
    )

    plt.figure(figsize=(figure_size, figure_size))

    image = plt.imshow(
        matrix_plot,
        aspect="auto",
        vmin=0,
        vmax=1 if normalize else None,
    )

    plt.colorbar(
        image,
        fraction=0.046,
        pad=0.04,
    )

    positions = np.arange(n_classes)

    plt.xticks(
        positions,
        labels,
        rotation=90,
        fontsize=8,
    )

    plt.yticks(
        positions,
        labels,
        fontsize=8,
    )

    plt.xlabel("Predicted room")
    plt.ylabel("True room")

    title_suffix = (
        "Normalized by true class"
        if normalize
        else "Sample count"
    )

    plt.title(
        f"Confusion Matrix ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â {test_day}\n"
        f"{protocol} | {stage} | {title_suffix}"
    )

    # ChÃƒÂ¡Ã‚Â»Ã¢â‚¬Â° annotate nÃƒÂ¡Ã‚ÂºÃ‚Â¿u sÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœ class khÃƒÆ’Ã‚Â´ng quÃƒÆ’Ã‚Â¡ lÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºn.
    if CF_ANNOTATE and n_classes <= CF_ANNOTATE_MAX_CLASSES:
        threshold = (
            0.50
            if normalize
            else (
                matrix_plot.max() * 0.50
                if matrix_plot.size and matrix_plot.max() > 0
                else 0
            )
        )

        for row_idx in range(n_classes):
            for col_idx in range(n_classes):
                value = matrix_plot[row_idx, col_idx]

                if normalize:
                    text = format(value, value_format)
                else:
                    text = format(int(value), value_format)

                plt.text(
                    col_idx,
                    row_idx,
                    text,
                    ha="center",
                    va="center",
                    fontsize=6,
                )

    plt.tight_layout()

    png_path = (
        output_dir
        / f"cf_{test_day}_{protocol}_{stage}_{matrix_type}.png"
    )

    plt.savefig(
        png_path,
        dpi=CF_FIG_DPI,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()

    return {
        "test_day": str(test_day),
        "protocol": protocol,
        "stage": stage,
        "matrix_type": matrix_type,
        "n_samples": int(len(y_true)),
        "n_classes": int(n_classes),
        "csv_path": str(csv_path),
        "png_path": str(png_path),
    }


# ------------------------------------------------------------
# Run all folds
# ------------------------------------------------------------
cf_manifest_rows = []

fold_dirs = sorted(
    [
        path
        for path in OUTPUT_DIR.glob("fold_*")
        if path.is_dir()
    ],
    key=lambda path: path.name,
)

if not fold_dirs:
    raise FileNotFoundError(
        f"KhÃƒÆ’Ã‚Â´ng tÃƒÆ’Ã‚Â¬m thÃƒÂ¡Ã‚ÂºÃ‚Â¥y folder fold_* trong:\n{OUTPUT_DIR}"
    )

for fold_dir in fold_dirs:
    prediction_file = fold_dir / "predictions.csv"

    if not prediction_file.exists():
        print("Skip: missing", prediction_file)
        continue

    predictions = pd.read_csv(prediction_file)

    required_columns = {"y_true"}

    missing_required = required_columns - set(predictions.columns)

    if missing_required:
        raise ValueError(
            f"{prediction_file} thiÃƒÂ¡Ã‚ÂºÃ‚Â¿u cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t: {sorted(missing_required)}"
        )

    test_day = fold_dir.name.replace("fold_", "", 1)

    # LÃƒÂ¡Ã‚ÂºÃ‚Â¥y stage tÃƒÂ¡Ã‚Â»Ã‚Â± Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢ng tÃƒÂ¡Ã‚Â»Ã‚Â« cÃƒÆ’Ã‚Â¡c cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t y_pred_<stage>.
    available_stage_columns = {
        column.replace("y_pred_", "", 1): column
        for column in predictions.columns
        if column.startswith("y_pred_")
    }

    if CF_STAGES is None:
        stages_to_plot = sorted(
            available_stage_columns.keys()
        )
    else:
        stages_to_plot = [
            stage
            for stage in CF_STAGES
            if stage in available_stage_columns
        ]

    if not stages_to_plot:
        print(
            f"Skip {test_day}: khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t y_pred_* phÃƒÆ’Ã‚Â¹ hÃƒÂ¡Ã‚Â»Ã‚Â£p."
        )
        continue

    print("=" * 100)
    print("CONFUSION MATRICES ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â TEST DAY:", test_day)
    print("Stages:", stages_to_plot)
    print("=" * 100)

    for protocol in CF_PROTOCOLS:
        protocol_mask = cf_get_protocol_mask(
            predictions,
            protocol,
        )

        protocol_df = predictions.loc[
            protocol_mask
        ].copy()

        if protocol_df.empty:
            print(
                f"Skip {test_day} / {protocol}: no samples"
            )
            continue

        protocol_df["y_true"] = (
            protocol_df["y_true"]
            .astype(str)
            .str.strip()
        )

        for stage in stages_to_plot:
            prediction_column = available_stage_columns[stage]

            stage_df = protocol_df[
                ["y_true", prediction_column]
            ].dropna().copy()

            stage_df[prediction_column] = (
                stage_df[prediction_column]
                .astype(str)
                .str.strip()
            )

            if stage_df.empty:
                print(
                    f"Skip {test_day} / {protocol} / {stage}: "
                    "no valid predictions"
                )
                continue

            y_true = stage_df["y_true"].to_numpy()
            y_pred = stage_df[prediction_column].to_numpy()

            labels = sorted(
                set(y_true) | set(y_pred),
                key=cf_sort_label_key,
            )

            for normalize in CF_NORMALIZATION_MODES:
                result = cf_plot_one_matrix(
                    y_true=y_true,
                    y_pred=y_pred,
                    labels=labels,
                    test_day=test_day,
                    stage=stage,
                    protocol=protocol,
                    normalize=normalize,
                    output_dir=CF_OUTPUT_DIR,
                )

                cf_manifest_rows.append(result)


# ------------------------------------------------------------
# Save manifest
# ------------------------------------------------------------
cf_manifest_df = pd.DataFrame(cf_manifest_rows)

cf_manifest_df.to_csv(
    CF_OUTPUT_DIR / "confusion_matrix_manifest.csv",
    index=False,
)

print("=" * 100)
print("CONFUSION MATRIX COMPLETED")
print("Output:", CF_OUTPUT_DIR)
print("Number of matrices:", len(cf_manifest_df))
print("=" * 100)

display(cf_manifest_df)
````

````python
# ============================================================
# CLOSED-SET CONFUSION MATRIX ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â RIÃƒÆ’Ã…Â NG THEO TÃƒÂ¡Ã‚Â»Ã‚ÂªNG NGÃƒÆ’Ã¢â€šÂ¬Y
#
# Input:
#   OUTPUT_DIR/fold_<date>/predictions.csv
#
# Output:
#   OUTPUT_DIR/closedset_confusion_by_day/
#
# MÃƒÂ¡Ã‚ÂºÃ‚Â·c Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¹nh:
#   - stage: viterbi_markov
#   - normalize theo true class
#   - mÃƒÂ¡Ã‚Â»Ã¢â‚¬â€i ngÃƒÆ’Ã‚Â y mÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t figure riÃƒÆ’Ã‚Âªng
#   - prediction sang class khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ trong test day Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c gom vÃƒÆ’Ã‚Â o OTHER
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix


# ------------------------------------------------------------
# Config
# ------------------------------------------------------------
CF_STAGE = "viterbi_markov"

# CÃƒÆ’Ã‚Â¡c stage cÃƒÆ’Ã‚Â³ thÃƒÂ¡Ã‚Â»Ã†â€™ dÃƒÆ’Ã‚Â¹ng:
# "argmax_raw"
# "prob_smooth"
# "viterbi_markov"

CF_NORMALIZE = True
CF_DPI = 220
CF_ANNOTATE = True

CF_OUTPUT_DIR = OUTPUT_DIR / "closedset_confusion_by_day"
CF_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# Helper
# ------------------------------------------------------------
def sort_room_label(label):
    """
    Room sÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœ Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚ÂºÃ‚Â·t trÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºc, semantic room Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚ÂºÃ‚Â·t sau.
    """
    label = str(label).strip()

    try:
        return 0, int(label)
    except ValueError:
        semantic_order = {
            "cafeteria": 1001,
            "cleaning": 1002,
            "hallway": 1003,
            "kitchen": 1004,
            "nurse station": 1005,
            "OTHER": 9999,
        }

        return 1, semantic_order.get(
            label.lower(),
            5000,
        ), label.lower()


def parse_boolean_column(series):
    """
    Ãƒâ€žÃ‚ÂÃƒÂ¡Ã‚Â»Ã‚Âc an toÃƒÆ’Ã‚Â n cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t is_unseen_test_class tÃƒÂ¡Ã‚Â»Ã‚Â« CSV.
    """
    if pd.api.types.is_bool_dtype(series):
        return series.to_numpy(dtype=bool)

    return (
        series.astype(str)
        .str.strip()
        .str.lower()
        .isin(["true", "1", "yes"])
        .to_numpy()
    )


def normalize_confusion_rows(matrix):
    matrix = np.asarray(matrix, dtype=float)

    row_sum = matrix.sum(
        axis=1,
        keepdims=True,
    )

    return np.divide(
        matrix,
        row_sum,
        out=np.zeros_like(matrix, dtype=float),
        where=row_sum > 0,
    )


def plot_closedset_confusion_one_day(
    predictions,
    test_day,
    prediction_column,
    stage,
):
    # --------------------------------------------------------
    # 1. Closed-set filter
    # --------------------------------------------------------
    if "is_unseen_test_class" not in predictions.columns:
        raise ValueError(
            f"{test_day}: thiÃƒÂ¡Ã‚ÂºÃ‚Â¿u cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t is_unseen_test_class."
        )

    unseen_mask = parse_boolean_column(
        predictions["is_unseen_test_class"]
    )

    closed_df = predictions.loc[
        ~unseen_mask,
        ["y_true", prediction_column],
    ].dropna().copy()

    if closed_df.empty:
        print(f"Skip {test_day}: khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ closed-set sample.")
        return None

    closed_df["y_true"] = (
        closed_df["y_true"]
        .astype(str)
        .str.strip()
    )

    closed_df[prediction_column] = (
        closed_df[prediction_column]
        .astype(str)
        .str.strip()
    )

    # --------------------------------------------------------
    # 2. ChÃƒÂ¡Ã‚Â»Ã¢â‚¬Â° dÃƒÆ’Ã‚Â¹ng class thÃƒÂ¡Ã‚Â»Ã‚Â±c sÃƒÂ¡Ã‚Â»Ã‚Â± xuÃƒÂ¡Ã‚ÂºÃ‚Â¥t hiÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡n trong test day
    # --------------------------------------------------------
    true_labels = sorted(
        closed_df["y_true"].unique(),
        key=sort_room_label,
    )

    true_label_set = set(true_labels)

    # Prediction sang class khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ support trong ngÃƒÆ’Ã‚Â y test
    # Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c gom thÃƒÆ’Ã‚Â nh OTHER Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã†â€™ khÃƒÆ’Ã‚Â´ng lÃƒÆ’Ã‚Â m mÃƒÂ¡Ã‚ÂºÃ‚Â¥t error count.
    has_other_prediction = (
        ~closed_df[prediction_column].isin(true_label_set)
    ).any()

    if has_other_prediction:
        closed_df["y_pred_plot"] = (
            closed_df[prediction_column]
            .where(
                closed_df[prediction_column].isin(
                    true_label_set
                ),
                "OTHER",
            )
        )

        plot_labels = true_labels + ["OTHER"]

    else:
        closed_df["y_pred_plot"] = (
            closed_df[prediction_column]
        )

        plot_labels = true_labels

    # --------------------------------------------------------
    # 3. Confusion matrix
    # --------------------------------------------------------
    matrix_count = confusion_matrix(
        closed_df["y_true"],
        closed_df["y_pred_plot"],
        labels=plot_labels,
    )

    # OTHER khÃƒÆ’Ã‚Â´ng phÃƒÂ¡Ã‚ÂºÃ‚Â£i true class nÃƒÆ’Ã‚Âªn hÃƒÆ’Ã‚Â ng OTHER sÃƒÂ¡Ã‚ÂºÃ‚Â½ bÃƒÂ¡Ã‚ÂºÃ‚Â±ng 0.
    # ChÃƒÂ¡Ã‚Â»Ã¢â‚¬Â° giÃƒÂ¡Ã‚Â»Ã‚Â¯ cÃƒÆ’Ã‚Â¡c hÃƒÆ’Ã‚Â ng true class khi hiÃƒÂ¡Ã‚Â»Ã†â€™n thÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¹.
    matrix_count_display = matrix_count[
        : len(true_labels),
        :,
    ]

    if CF_NORMALIZE:
        matrix_display = normalize_confusion_rows(
            matrix_count_display
        )
    else:
        matrix_display = matrix_count_display

    # --------------------------------------------------------
    # 4. Support tÃƒÂ¡Ã‚Â»Ã‚Â«ng true class
    # --------------------------------------------------------
    supports = (
        closed_df["y_true"]
        .value_counts()
        .reindex(true_labels)
        .fillna(0)
        .astype(int)
    )

    y_tick_labels = [
        f"{label} (n={supports[label]})"
        for label in true_labels
    ]

    # --------------------------------------------------------
    # 5. Save CSV
    # --------------------------------------------------------
    matrix_count_df = pd.DataFrame(
        matrix_count_display,
        index=true_labels,
        columns=plot_labels,
    )

    matrix_count_df.index.name = "true_room"
    matrix_count_df.columns.name = "predicted_room"

    matrix_count_df.to_csv(
        CF_OUTPUT_DIR
        / f"closedset_cf_{test_day}_{stage}_count.csv"
    )

    matrix_normalized_df = pd.DataFrame(
        normalize_confusion_rows(
            matrix_count_display
        ),
        index=true_labels,
        columns=plot_labels,
    )

    matrix_normalized_df.index.name = "true_room"
    matrix_normalized_df.columns.name = "predicted_room"

    matrix_normalized_df.to_csv(
        CF_OUTPUT_DIR
        / f"closedset_cf_{test_day}_{stage}_normalized.csv"
    )

    # --------------------------------------------------------
    # 6. Plot
    # --------------------------------------------------------
    n_rows = len(true_labels)
    n_cols = len(plot_labels)

    width = max(
        9,
        min(18, n_cols * 0.85),
    )

    height = max(
        7,
        min(18, n_rows * 0.75),
    )

    plt.figure(figsize=(width, height))

    image = plt.imshow(
        matrix_display,
        aspect="auto",
        vmin=0,
        vmax=1 if CF_NORMALIZE else None,
    )

    colorbar = plt.colorbar(
        image,
        fraction=0.046,
        pad=0.04,
    )

    colorbar.set_label(
        "Recall proportion"
        if CF_NORMALIZE
        else "Number of samples"
    )

    plt.xticks(
        np.arange(n_cols),
        plot_labels,
        rotation=45,
        ha="right",
        fontsize=10,
    )

    plt.yticks(
        np.arange(n_rows),
        y_tick_labels,
        fontsize=10,
    )

    plt.xlabel("Predicted room")
    plt.ylabel("True room")

    metric_text = (
        "Normalized by true class"
        if CF_NORMALIZE
        else "Sample count"
    )

    plt.title(
        f"Closed-set Confusion Matrix ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â {test_day}\n"
        f"Stage: {stage} | {metric_text}"
    )

    # --------------------------------------------------------
    # 7. Cell annotation
    # --------------------------------------------------------
    if CF_ANNOTATE:
        for row_idx in range(n_rows):
            for col_idx in range(n_cols):
                value = matrix_display[
                    row_idx,
                    col_idx,
                ]

                count_value = matrix_count_display[
                    row_idx,
                    col_idx,
                ]

                if CF_NORMALIZE:
                    if count_value == 0:
                        text = ""
                    else:
                        text = (
                            f"{value:.2f}\n"
                            f"({count_value})"
                        )
                else:
                    text = (
                        str(int(count_value))
                        if count_value > 0
                        else ""
                    )

                if text:
                    plt.text(
                        col_idx,
                        row_idx,
                        text,
                        ha="center",
                        va="center",
                        fontsize=8,
                    )

    plt.tight_layout()

    output_png = (
        CF_OUTPUT_DIR
        / f"closedset_cf_{test_day}_{stage}.png"
    )

    plt.savefig(
        output_png,
        dpi=CF_DPI,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()

    # --------------------------------------------------------
    # 8. Summary
    # --------------------------------------------------------
    correct = int(
        (
            closed_df["y_true"]
            == closed_df[prediction_column]
        ).sum()
    )

    summary = {
        "test_day": str(test_day),
        "stage": stage,
        "n_closedset_samples": int(
            len(closed_df)
        ),
        "n_true_classes": int(
            len(true_labels)
        ),
        "accuracy": float(
            correct / len(closed_df)
        ),
        "has_other_prediction": bool(
            has_other_prediction
        ),
        "output_png": str(output_png),
    }

    return summary


# ------------------------------------------------------------
# Run tÃƒÂ¡Ã‚Â»Ã‚Â«ng ngÃƒÆ’Ã‚Â y
# ------------------------------------------------------------
fold_dirs = sorted(
    [
        path
        for path in OUTPUT_DIR.glob("fold_*")
        if path.is_dir()
    ],
    key=lambda path: path.name,
)

if not fold_dirs:
    raise FileNotFoundError(
        f"KhÃƒÆ’Ã‚Â´ng tÃƒÆ’Ã‚Â¬m thÃƒÂ¡Ã‚ÂºÃ‚Â¥y fold_* trong {OUTPUT_DIR}"
    )

summary_rows = []

for fold_dir in fold_dirs:
    test_day = fold_dir.name.replace(
        "fold_",
        "",
        1,
    )

    prediction_file = (
        fold_dir / "predictions.csv"
    )

    if not prediction_file.exists():
        print(
            "Skip, missing:",
            prediction_file,
        )
        continue

    predictions = pd.read_csv(
        prediction_file
    )

    preferred_column = f"y_pred_{CF_STAGE}"

    # Fallback nÃƒÂ¡Ã‚ÂºÃ‚Â¿u stage Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c chÃƒÂ¡Ã‚Â»Ã‚Ân khÃƒÆ’Ã‚Â´ng tÃƒÂ¡Ã‚Â»Ã¢â‚¬Å“n tÃƒÂ¡Ã‚ÂºÃ‚Â¡i.
    if preferred_column in predictions.columns:
        prediction_column = preferred_column
        stage_used = CF_STAGE

    else:
        available_columns = [
            c
            for c in predictions.columns
            if c.startswith("y_pred_")
        ]

        if not available_columns:
            print(
                f"Skip {test_day}: khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ y_pred_*."
            )
            continue

        prediction_column = available_columns[-1]
        stage_used = prediction_column.replace(
            "y_pred_",
            "",
            1,
        )

        print(
            f"{test_day}: khÃƒÆ’Ã‚Â´ng cÃƒÆ’Ã‚Â³ {preferred_column}; "
            f"dÃƒÆ’Ã‚Â¹ng {prediction_column}."
        )

    print("=" * 90)
    print("TEST DAY:", test_day)
    print("STAGE   :", stage_used)
    print("=" * 90)

    result = plot_closedset_confusion_one_day(
        predictions=predictions,
        test_day=test_day,
        prediction_column=prediction_column,
        stage=stage_used,
    )

    if result is not None:
        summary_rows.append(result)


# ------------------------------------------------------------
# Save summary
# ------------------------------------------------------------
closedset_cf_summary = pd.DataFrame(
    summary_rows
)

closedset_cf_summary.to_csv(
    CF_OUTPUT_DIR
    / "closedset_confusion_summary.csv",
    index=False,
)

print("=" * 90)
print("COMPLETED")
print("Output:", CF_OUTPUT_DIR)
print("=" * 90)

display(closedset_cf_summary)
````
