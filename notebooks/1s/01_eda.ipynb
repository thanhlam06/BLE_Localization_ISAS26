````python
# ============================================================
# 1. LOAD DATA
# ============================================================

def resolve_feature_file() -> Path:
    if FEATURE_FILE.exists():
        return FEATURE_FILE

    for p in FALLBACK_FILES:
        if p.exists():
            return p

    raise FileNotFoundError(
        "KhÃƒÆ’Ã‚Â´ng tÃƒÆ’Ã‚Â¬m thÃƒÂ¡Ã‚ÂºÃ‚Â¥y rf_window_features_1s.csv. "
        f"Ãƒâ€žÃ‚ÂÃƒÆ’Ã‚Â£ kiÃƒÂ¡Ã‚Â»Ã†â€™m tra: {[str(FEATURE_FILE)] + [str(p) for p in FALLBACK_FILES]}"
    )


def normalize_date_and_target(data: pd.DataFrame) -> pd.DataFrame:
    data = data.copy()

    if TIME_COL in data.columns:
        data[TIME_COL] = pd.to_datetime(data[TIME_COL], errors="coerce")

    if DATE_COL in data.columns:
        parsed_date = pd.to_datetime(data[DATE_COL], errors="coerce")
        data[DATE_COL] = parsed_date.dt.date.astype("string")
    elif TIME_COL in data.columns:
        data[DATE_COL] = data[TIME_COL].dt.date.astype("string")
    else:
        raise ValueError(f"ThiÃƒÂ¡Ã‚ÂºÃ‚Â¿u cÃƒÂ¡Ã‚ÂºÃ‚Â£ {DATE_COL!r} vÃƒÆ’Ã‚Â  {TIME_COL!r}.")

    if TARGET_COL not in data.columns:
        raise ValueError(f"ThiÃƒÂ¡Ã‚ÂºÃ‚Â¿u target column: {TARGET_COL}")

    data[TARGET_COL] = data[TARGET_COL].astype("string").str.strip()
    data = data.dropna(subset=[DATE_COL, TARGET_COL]).reset_index(drop=True)

    return data


resolved_file = resolve_feature_file()
df = pd.read_csv(resolved_file)
df = normalize_date_and_target(df)

days = sorted(df[DATE_COL].dropna().astype(str).unique().tolist())
rooms = sorted(df[TARGET_COL].dropna().astype(str).unique().tolist())

print("Loaded:", resolved_file)
print("Shape:", df.shape)
print("Days:", days)
print("Number of rooms:", len(rooms))
print("Rooms:", rooms)
````

````python
# ============================================================
# 3. BLE-SAFE FEATURE SELECTION
# ============================================================

DIAGNOSTIC_PATTERNS = [
    "label_reliability", "label_confidence", "label_duration",
    "boundary_distance", "boundary_ratio", "is_boundary",
    "interval", "annotation", "overlap",
    "groundtruth", "ground_truth", "gt_",
]

TIME_FEATURES = {
    "hour", "minute", "second", "dayofweek",
    "time_sin", "time_cos",
}

METADATA_COLS = {
    TARGET_COL, DATE_COL, TIME_COL,
    "timestamp", "window_start", "window_end",
    "start_time", "end_time",
    "day_group", "fold", "test_day",
    "y_true", "y_pred", "_source_file",
}


def contains_inactive_beacon(name: str) -> bool:
    s = str(name).lower()
    return bool(re.search(r"(?:^|_)b(?:24|25)(?:_|$)", s))


def is_diagnostic_feature(name: str) -> bool:
    s = str(name).lower()
    return any(pattern in s for pattern in DIAGNOSTIC_PATTERNS)


def is_id_feature(name: str) -> bool:
    s = str(name).lower()
    return s.endswith("_id") or "_id_" in s


def is_discrete_feature(name: str) -> bool:
    s = str(name).lower()
    return (
        s.startswith("present_b")
        or is_id_feature(s)
        or s.startswith("top1_beacon")
        or s.startswith("top2_beacon")
        or s.startswith("top3_beacon")
    )


def build_ble_safe_features(data: pd.DataFrame):
    selected = []
    audit_rows = []

    for c in data.columns:
        reason = None

        if c in METADATA_COLS:
            reason = "metadata"
        elif c in TIME_FEATURES:
            reason = "time_feature"
        elif is_diagnostic_feature(c):
            reason = "diagnostic"
        elif contains_inactive_beacon(c):
            reason = "inactive_beacon_B24_B25"
        elif str(c).lower().startswith("power"):
            reason = "power_feature"
        elif not pd.api.types.is_numeric_dtype(data[c]):
            reason = "non_numeric"
        elif data[c].nunique(dropna=True) <= 1:
            reason = "constant"
        else:
            selected.append(c)
            reason = "selected"

        audit_rows.append({
            "feature": c,
            "status": reason,
            "dtype": str(data[c].dtype),
            "n_unique": int(data[c].nunique(dropna=True)),
            "missing_ratio": float(data[c].isna().mean()),
        })

    return selected, pd.DataFrame(audit_rows)


feature_cols, feature_audit = build_ble_safe_features(df)

# PCA/correlation/boxplot khÃƒÆ’Ã‚Â´ng dÃƒÆ’Ã‚Â¹ng categorical beacon ID nhÃƒâ€ Ã‚Â° continuous values.
continuous_feature_cols = [c for c in feature_cols if not is_id_feature(c)]
discrete_mask_all = np.array([is_discrete_feature(c) for c in feature_cols], dtype=bool)

feature_audit.to_csv(OUT_DIR / "feature_selection_audit.csv", index=False)

print("BLE-safe features:", len(feature_cols))
print("Continuous features:", len(continuous_feature_cols))
print("Discrete/categorical-coded features:", int(discrete_mask_all.sum()))
display(feature_audit["status"].value_counts().rename_axis("status").reset_index(name="count"))
display(feature_audit[feature_audit["status"] == "selected"].head(30))
````

````python
# ============================================================
# 4. SAMPLING HELPERS
# ============================================================

def balanced_sample_by_day(
    data: pd.DataFrame,
    max_rows: int,
    random_state: int = RANDOM_STATE,
) -> pd.DataFrame:
    if len(data) <= max_rows:
        return data.copy().reset_index(drop=True)

    day_values = sorted(data[DATE_COL].astype(str).unique())
    per_day = max(1, max_rows // len(day_values))

    selected_indices = []
    for offset, day in enumerate(day_values):
        part = data[data[DATE_COL].astype(str) == day]
        n_take = min(len(part), per_day)
        selected_indices.extend(
            part.sample(
                n=n_take,
                random_state=random_state + offset,
            ).index.tolist()
        )

    selected_indices = list(dict.fromkeys(selected_indices))

    if len(selected_indices) < max_rows:
        remaining_indices = data.index.difference(selected_indices)
        extra_n = min(max_rows - len(selected_indices), len(remaining_indices))
        if extra_n > 0:
            extra = data.loc[remaining_indices].sample(
                n=extra_n,
                random_state=random_state + 100,
            )
            selected_indices.extend(extra.index.tolist())

    sampled = data.loc[selected_indices]
    return sampled.sample(
        frac=1.0,
        random_state=random_state,
    ).reset_index(drop=True)


def save_current_figure(path: Path):
    plt.tight_layout()
    plt.savefig(path, dpi=220, bbox_inches="tight")
    plt.show()
    plt.close()
````

## 5. PCA

PCA Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c fit trÃƒÆ’Ã‚Âªn feature BLE-safe liÃƒÆ’Ã‚Âªn tÃƒÂ¡Ã‚Â»Ã‚Â¥c sau:

```text
median imputation ÃƒÂ¢Ã¢â‚¬Â Ã¢â‚¬â„¢ StandardScaler ÃƒÂ¢Ã¢â‚¬Â Ã¢â‚¬â„¢ PCA
```

Standardization lÃƒÆ’Ã‚Â  bÃƒÂ¡Ã‚ÂºÃ‚Â¯t buÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢c ÃƒÂ¡Ã‚Â»Ã…Â¸ Ãƒâ€žÃ¢â‚¬ËœÃƒÆ’Ã‚Â¢y vÃƒÆ’Ã‚Â¬ PCA phÃƒÂ¡Ã‚Â»Ã‚Â¥ thuÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢c phÃƒâ€ Ã‚Â°Ãƒâ€ Ã‚Â¡ng sai vÃƒÆ’Ã‚Â  thang Ãƒâ€žÃ¢â‚¬Ëœo feature.


````python
# ============================================================
# 5. PCA
# ============================================================

pca_df = balanced_sample_by_day(
    df[[DATE_COL, TARGET_COL] + continuous_feature_cols],
    max_rows=PCA_SAMPLE_MAX,
)

pca_imputer = SimpleImputer(strategy="median")
pca_scaler = StandardScaler()

X_pca_raw = pca_df[continuous_feature_cols]
X_pca_imputed = pca_imputer.fit_transform(X_pca_raw)
X_pca_scaled = pca_scaler.fit_transform(X_pca_imputed)

n_components = min(
    PCA_N_COMPONENTS,
    X_pca_scaled.shape[0] - 1,
    X_pca_scaled.shape[1],
)

pca = PCA(
    n_components=n_components,
    svd_solver="randomized",
    random_state=RANDOM_STATE,
)
pca_scores = pca.fit_transform(X_pca_scaled)

explained_df = pd.DataFrame({
    "component": [f"PC{i+1}" for i in range(n_components)],
    "explained_variance_ratio": pca.explained_variance_ratio_,
    "cumulative_explained_variance": np.cumsum(pca.explained_variance_ratio_),
})
explained_df.to_csv(OUT_DIR / "pca_explained_variance.csv", index=False)

loadings_df = pd.DataFrame(
    pca.components_.T,
    index=continuous_feature_cols,
    columns=[f"PC{i+1}" for i in range(n_components)],
)
loadings_df.index.name = "feature"
loadings_df.to_csv(OUT_DIR / "pca_loadings.csv")

score_cols = [f"PC{i+1}" for i in range(n_components)]
pca_scores_df = pd.DataFrame(pca_scores, columns=score_cols)
pca_scores_df[DATE_COL] = pca_df[DATE_COL].astype(str).to_numpy()
pca_scores_df[TARGET_COL] = pca_df[TARGET_COL].astype(str).to_numpy()
pca_scores_df.to_csv(OUT_DIR / "pca_scores_sample.csv", index=False)

# Daily PCA centroids
pca_centroids = (
    pca_scores_df.groupby(DATE_COL)[score_cols[:min(10, n_components)]]
                 .mean()
                 .reset_index()
)
pca_centroids.to_csv(OUT_DIR / "pca_daily_centroids.csv", index=False)

# Pairwise Euclidean distance between daily centroids
centroid_matrix = pca_centroids.set_index(DATE_COL)
centroid_rows = []
centroid_days = centroid_matrix.index.tolist()
for i, day_a in enumerate(centroid_days):
    for day_b in centroid_days[i + 1:]:
        dist = float(np.linalg.norm(
            centroid_matrix.loc[day_a].to_numpy()
            - centroid_matrix.loc[day_b].to_numpy()
        ))
        centroid_rows.append({
            "day_a": day_a,
            "day_b": day_b,
            "pca_centroid_distance": dist,
        })

pca_centroid_distances = pd.DataFrame(centroid_rows)
pca_centroid_distances.to_csv(
    OUT_DIR / "pca_daily_centroid_pairwise_distances.csv",
    index=False,
)

# Plot cumulative explained variance
plt.figure(figsize=(9, 5))
plt.plot(
    np.arange(1, n_components + 1),
    explained_df["cumulative_explained_variance"],
    marker="o",
)
plt.xlabel("Number of principal components")
plt.ylabel("Cumulative explained variance ratio")
plt.title("PCA cumulative explained variance ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â 1s BLE features")
plt.grid(True, alpha=0.3)
save_current_figure(PLOT_DIR / "pca_cumulative_explained_variance.png")

# Plot PC1 vs PC2 by day
if n_components >= 2:
    plt.figure(figsize=(10, 7))
    for day in sorted(pca_scores_df[DATE_COL].unique()):
        part = pca_scores_df[pca_scores_df[DATE_COL] == day]
        plt.scatter(
            part["PC1"],
            part["PC2"],
            s=10,
            alpha=0.35,
            label=day,
        )
    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.title("PCA projection by day ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â 1s BLE features")
    plt.legend(title="Day")
    plt.grid(True, alpha=0.3)
    save_current_figure(PLOT_DIR / "pca_pc1_pc2_by_day.png")

display(explained_df.head(10))
display(pca_centroid_distances)
display(
    loadings_df.assign(
        abs_PC1=loadings_df["PC1"].abs()
    ).sort_values("abs_PC1", ascending=False).head(20)
)
````

## 6. Correlation matrix

Notebook lÃƒâ€ Ã‚Â°u:

- ma trÃƒÂ¡Ã‚ÂºÃ‚Â­n tÃƒâ€ Ã‚Â°Ãƒâ€ Ã‚Â¡ng quan Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚ÂºÃ‚Â§y Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã‚Â§;
- danh sÃƒÆ’Ã‚Â¡ch cÃƒÂ¡Ã‚ÂºÃ‚Â·p feature cÃƒÆ’Ã‚Â³ \(|r| \geq\) `HIGH_CORR_THRESHOLD`;
- heatmap cho cÃƒÆ’Ã‚Â¡c feature cÃƒÆ’Ã‚Â³ phÃƒâ€ Ã‚Â°Ãƒâ€ Ã‚Â¡ng sai lÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºn nhÃƒÂ¡Ã‚ÂºÃ‚Â¥t sau median imputation.


````python
# ============================================================
# 6. CORRELATION MATRIX
# ============================================================

corr_df = balanced_sample_by_day(
    df[[DATE_COL] + continuous_feature_cols],
    max_rows=CORR_SAMPLE_MAX,
)

corr_imputer = SimpleImputer(strategy="median")
X_corr = corr_imputer.fit_transform(corr_df[continuous_feature_cols])
X_corr_df = pd.DataFrame(X_corr, columns=continuous_feature_cols)

full_corr = X_corr_df.corr(method="pearson")
full_corr.to_csv(OUT_DIR / "correlation_matrix_full.csv")

# Extract upper-triangle highly correlated pairs
corr_values = full_corr.to_numpy()
corr_features = full_corr.columns.to_list()
high_corr_rows = []

for i in range(len(corr_features)):
    for j in range(i + 1, len(corr_features)):
        r = corr_values[i, j]
        if np.isfinite(r) and abs(r) >= HIGH_CORR_THRESHOLD:
            high_corr_rows.append({
                "feature_a": corr_features[i],
                "feature_b": corr_features[j],
                "pearson_r": float(r),
                "abs_pearson_r": float(abs(r)),
            })

high_corr_pairs = pd.DataFrame(high_corr_rows)
if not high_corr_pairs.empty:
    high_corr_pairs = high_corr_pairs.sort_values(
        "abs_pearson_r", ascending=False
    )
high_corr_pairs.to_csv(OUT_DIR / "highly_correlated_feature_pairs.csv", index=False)

# Heatmap: top-variance features
feature_variances = X_corr_df.var(axis=0).sort_values(ascending=False)
heatmap_features = feature_variances.head(
    min(CORR_HEATMAP_TOP_N, len(feature_variances))
).index.tolist()
heatmap_corr = full_corr.loc[heatmap_features, heatmap_features]

fig_size = max(10, min(22, len(heatmap_features) * 0.45))
plt.figure(figsize=(fig_size, fig_size))
image = plt.imshow(heatmap_corr.to_numpy(), aspect="auto", vmin=-1, vmax=1)
plt.colorbar(image, fraction=0.046, pad=0.04)
plt.xticks(
    np.arange(len(heatmap_features)),
    heatmap_features,
    rotation=90,
    fontsize=7,
)
plt.yticks(
    np.arange(len(heatmap_features)),
    heatmap_features,
    fontsize=7,
)
plt.title(f"Pearson correlation ÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Â top {len(heatmap_features)} variance features")
save_current_figure(PLOT_DIR / "correlation_heatmap_top_variance_features.png")

print("Full correlation shape:", full_corr.shape)
print("High-correlation pairs:", len(high_corr_pairs))
display(high_corr_pairs.head(30))
````

## 8. Chi-square test theo ngÃƒÆ’Ã‚Â y

MÃƒÂ¡Ã‚Â»Ã‚Â¥c tiÃƒÆ’Ã‚Âªu lÃƒÆ’Ã‚Â  kiÃƒÂ¡Ã‚Â»Ã†â€™m tra feature cÃƒÆ’Ã‚Â³ Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢c lÃƒÂ¡Ã‚ÂºÃ‚Â­p vÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºi ngÃƒÆ’Ã‚Â y hay khÃƒÆ’Ã‚Â´ng.

- Feature liÃƒÆ’Ã‚Âªn tÃƒÂ¡Ã‚Â»Ã‚Â¥c: chia thÃƒÆ’Ã‚Â nh cÃƒÆ’Ã‚Â¡c quantile bin.
- Feature nhÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¹ phÃƒÆ’Ã‚Â¢n/categorical-coded: giÃƒÂ¡Ã‚Â»Ã‚Â¯ category gÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœc.
- Missing value Ãƒâ€žÃ¢â‚¬ËœÃƒâ€ Ã‚Â°ÃƒÂ¡Ã‚Â»Ã‚Â£c giÃƒÂ¡Ã‚Â»Ã‚Â¯ thÃƒÆ’Ã‚Â nh mÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t category riÃƒÆ’Ã‚Âªng.
- `q_value_bh` lÃƒÆ’Ã‚Â  p-value sau BenjaminiÃƒÂ¢Ã¢â€šÂ¬Ã¢â‚¬Å“Hochberg.
- CramÃƒÆ’Ã‚Â©r's V Ãƒâ€žÃ¢â‚¬Ëœo mÃƒÂ¡Ã‚Â»Ã‚Â©c Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢ liÃƒÆ’Ã‚Âªn hÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡, khÃƒÆ’Ã‚Â´ng chÃƒÂ¡Ã‚Â»Ã¢â‚¬Â° ÃƒÆ’Ã‚Â½ nghÃƒâ€žÃ‚Â©a thÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœng kÃƒÆ’Ã‚Âª.

VÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºi dÃƒÂ¡Ã‚Â»Ã‚Â¯ liÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡u lÃƒÂ¡Ã‚Â»Ã¢â‚¬Âºn, p-value cÃƒÆ’Ã‚Â³ thÃƒÂ¡Ã‚Â»Ã†â€™ rÃƒÂ¡Ã‚ÂºÃ‚Â¥t nhÃƒÂ¡Ã‚Â»Ã‚Â ngay cÃƒÂ¡Ã‚ÂºÃ‚Â£ khi hiÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡u ÃƒÂ¡Ã‚Â»Ã‚Â©ng yÃƒÂ¡Ã‚ÂºÃ‚Â¿u; cÃƒÂ¡Ã‚ÂºÃ‚Â§n Ãƒâ€ Ã‚Â°u tiÃƒÆ’Ã‚Âªn CramÃƒÆ’Ã‚Â©r's V vÃƒÆ’Ã‚Â  MI(day).


````python
# ============================================================
# 8. CHI-SQUARE TEST: FEATURE VS DAY
# ============================================================

def benjamini_hochberg(p_values):
    p = np.asarray(p_values, dtype=float)
    n = len(p)

    order = np.argsort(p)
    ranked = p[order]

    adjusted = np.empty(n, dtype=float)
    running_min = 1.0

    for i in range(n - 1, -1, -1):
        rank = i + 1
        candidate = ranked[i] * n / rank
        running_min = min(running_min, candidate)
        adjusted[i] = min(running_min, 1.0)

    result = np.empty(n, dtype=float)
    result[order] = adjusted
    return result


def discretize_for_chi_square(
    series: pd.Series,
    feature_name: str,
    n_bins: int = CHI_SQUARE_BINS,
) -> pd.Series:
    s = pd.to_numeric(series, errors="coerce")

    # Category riÃƒÆ’Ã‚Âªng cho missing.
    missing_mask = s.isna()

    # Binary/discrete-coded features: dÃƒÆ’Ã‚Â¹ng category gÃƒÂ¡Ã‚Â»Ã¢â‚¬Ëœc.
    if is_discrete_feature(feature_name) or s.nunique(dropna=True) <= n_bins:
        out = s.astype("object")
        out[missing_mask] = "MISSING"
        return out.astype(str)

    # Continuous feature: quantile bins.
    try:
        binned = pd.qcut(
            s,
            q=n_bins,
            duplicates="drop",
        )
        out = binned.astype("object")
        out[missing_mask] = "MISSING"
        return out.astype(str)
    except Exception:
        out = s.round(6).astype("object")
        out[missing_mask] = "MISSING"
        return out.astype(str)


chi_df = balanced_sample_by_day(
    df[[DATE_COL] + feature_cols],
    max_rows=CHI_SAMPLE_MAX,
)

chi_rows = []
day_series = chi_df[DATE_COL].astype(str)

for feature in feature_cols:
    binned_feature = discretize_for_chi_square(chi_df[feature], feature)
    contingency = pd.crosstab(day_series, binned_feature, dropna=False)

    # CÃƒÂ¡Ã‚ÂºÃ‚Â§n ÃƒÆ’Ã‚Â­t nhÃƒÂ¡Ã‚ÂºÃ‚Â¥t 2 hÃƒÆ’Ã‚Â ng vÃƒÆ’Ã‚Â  2 cÃƒÂ¡Ã‚Â»Ã¢â€žÂ¢t.
    if contingency.shape[0] < 2 or contingency.shape[1] < 2:
        chi_rows.append({
            "feature": feature,
            "chi2": np.nan,
            "dof": np.nan,
            "p_value": np.nan,
            "cramers_v": np.nan,
            "n_categories": contingency.shape[1],
            "min_expected_count": np.nan,
            "valid_test": False,
        })
        continue

    chi2_stat, p_value, dof, expected = chi2_contingency(
        contingency.to_numpy(),
        correction=False,
    )

    n = contingency.to_numpy().sum()
    r, k = contingency.shape
    denominator = max(1, min(r - 1, k - 1))
    cramers_v = math.sqrt((chi2_stat / n) / denominator)

    chi_rows.append({
        "feature": feature,
        "chi2": float(chi2_stat),
        "dof": int(dof),
        "p_value": float(p_value),
        "cramers_v": float(cramers_v),
        "n_categories": int(contingency.shape[1]),
        "min_expected_count": float(np.min(expected)),
        "valid_test": True,
    })

chi_results = pd.DataFrame(chi_rows)

valid_mask = chi_results["valid_test"] & chi_results["p_value"].notna()
chi_results["q_value_bh"] = np.nan
chi_results.loc[valid_mask, "q_value_bh"] = benjamini_hochberg(
    chi_results.loc[valid_mask, "p_value"].to_numpy()
)
chi_results["significant_bh"] = chi_results["q_value_bh"] < ALPHA

chi_results = chi_results.sort_values(
    ["cramers_v", "q_value_bh"],
    ascending=[False, True],
)
chi_results.to_csv(
    OUT_DIR / "chi_square_feature_vs_day.csv",
    index=False,
)

top_chi = (
    chi_results.dropna(subset=["cramers_v"])
               .nlargest(CHI_TOP_N, "cramers_v")
               .sort_values("cramers_v")
)

plt.figure(figsize=(10, max(6, len(top_chi) * 0.28)))
plt.barh(top_chi["feature"], top_chi["cramers_v"])
plt.xlabel("CramÃƒÆ’Ã‚Â©r's V")
plt.ylabel("Feature")
plt.title(f"Top {len(top_chi)} day-dependent features by Chi-square")
save_current_figure(PLOT_DIR / "chi_square_top_day_dependent_features.png")

display(chi_results.head(40))
````

## 9. Box plot so sÃƒÆ’Ã‚Â¡nh feature giÃƒÂ¡Ã‚Â»Ã‚Â¯a cÃƒÆ’Ã‚Â¡c ngÃƒÆ’Ã‚Â y

NÃƒÂ¡Ã‚ÂºÃ‚Â¿u `BOXPLOT_FEATURES = None`, notebook tÃƒÂ¡Ã‚Â»Ã‚Â± chÃƒÂ¡Ã‚Â»Ã‚Ân feature liÃƒÆ’Ã‚Âªn tÃƒÂ¡Ã‚Â»Ã‚Â¥c cÃƒÆ’Ã‚Â³ CramÃƒÆ’Ã‚Â©r's V cao nhÃƒÂ¡Ã‚ÂºÃ‚Â¥t.  
CÃƒÆ’Ã‚Â¡c feature ID categorical bÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¹ loÃƒÂ¡Ã‚ÂºÃ‚Â¡i khÃƒÂ¡Ã‚Â»Ã‚Âi box plot.


````python
# ============================================================
# 9. BOX PLOTS BY DAY
# ============================================================

def select_boxplot_features():
    if BOXPLOT_FEATURES is not None:
        valid = [
            c for c in BOXPLOT_FEATURES
            if c in continuous_feature_cols
        ]
        if not valid:
            raise ValueError(
                "BOXPLOT_FEATURES khÃƒÆ’Ã‚Â´ng chÃƒÂ¡Ã‚Â»Ã‚Â©a feature liÃƒÆ’Ã‚Âªn tÃƒÂ¡Ã‚Â»Ã‚Â¥c hÃƒÂ¡Ã‚Â»Ã‚Â£p lÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡."
            )
        return valid

    candidates = chi_results[
        chi_results["feature"].isin(continuous_feature_cols)
        & chi_results["cramers_v"].notna()
    ].copy()

    selected = (
        candidates.nlargest(BOXPLOT_TOP_N, "cramers_v")["feature"]
                  .tolist()
    )

    # Fallback bÃƒÂ¡Ã‚ÂºÃ‚Â±ng MI(day) nÃƒÂ¡Ã‚ÂºÃ‚Â¿u Chi-square khÃƒÆ’Ã‚Â´ng Ãƒâ€žÃ¢â‚¬ËœÃƒÂ¡Ã‚Â»Ã‚Â§ feature hÃƒÂ¡Ã‚Â»Ã‚Â£p lÃƒÂ¡Ã‚Â»Ã¢â‚¬Â¡.
    if len(selected) < BOXPLOT_TOP_N:
        mi_fallback = (
            mi_results[
                mi_results["feature"].isin(continuous_feature_cols)
                & ~mi_results["feature"].isin(selected)
            ]
            .nlargest(BOXPLOT_TOP_N - len(selected), "mi_day")["feature"]
            .tolist()
        )
        selected.extend(mi_fallback)

    return selected


boxplot_features = select_boxplot_features()

boxplot_stats_rows = []

for feature in boxplot_features:
    values_by_day = []

    for day in days:
        values = pd.to_numeric(
            df.loc[df[DATE_COL].astype(str) == day, feature],
            errors="coerce",
        ).dropna()

        values_by_day.append(values.to_numpy())

        if len(values) > 0:
            q1 = float(values.quantile(0.25))
            median = float(values.quantile(0.50))
            q3 = float(values.quantile(0.75))
            mean = float(values.mean())
            std = float(values.std())
        else:
            q1 = median = q3 = mean = std = np.nan

        total_day_rows = int((df[DATE_COL].astype(str) == day).sum())

        boxplot_stats_rows.append({
            "feature": feature,
            "day": day,
            "n_valid": int(len(values)),
            "n_total": total_day_rows,
            "missing_ratio": float(1.0 - len(values) / max(total_day_rows, 1)),
            "mean": mean,
            "std": std,
            "q1": q1,
            "median": median,
            "q3": q3,
            "iqr": q3 - q1 if np.isfinite(q1) and np.isfinite(q3) else np.nan,
        })

    plt.figure(figsize=(10, 6))
    plt.boxplot(
        values_by_day,
        labels=days,
        showfliers=BOXPLOT_SHOW_FLIERS,
    )
    plt.xlabel("Day")
    plt.ylabel(feature)
    plt.title(f"{feature} distribution by day")
    plt.grid(True, axis="y", alpha=0.3)

    safe_name = re.sub(r"[^A-Za-z0-9_.-]+", "_", feature)
    save_current_figure(BOXPLOT_DIR / f"boxplot_{safe_name}.png")

boxplot_stats = pd.DataFrame(boxplot_stats_rows)
boxplot_stats.to_csv(
    OUT_DIR / "boxplot_selected_features_daily_statistics.csv",
    index=False,
)

pd.DataFrame({"selected_boxplot_feature": boxplot_features}).to_csv(
    OUT_DIR / "boxplot_selected_features.csv",
    index=False,
)

print("Selected boxplot features:")
for feature in boxplot_features:
    print(" -", feature)

display(boxplot_stats.head(40))
````
